```python
import subprocess
import os
import tempfile
import re

# --- Configuration Paths ---
MXFOLD2_CONFIG = "@/kaggle/input/datasets/khushvaghela/mxfold2/models-0.1.0/models/TrainSetA.conf"
IPKNOT_BIN = "/kaggle/working/ipknot/build/ipknot"
VIENNA_LIB_PATH = "/usr/local/lib/python3.12/dist-packages/viennarna.libs"

def run_mxfold2(sequence):
    """Runs MxFold2 and returns the dot-bracket string."""
    with tempfile.NamedTemporaryFile(mode='w', suffix='.fa', delete=True) as tmp:
        tmp.write(f">temp\n{sequence}\n")
        tmp.flush()
        
        cmd = ["mxfold2", "predict", MXFOLD2_CONFIG, tmp.name]
        result = subprocess.run(cmd, capture_output=True, text=True)
        
        if result.returncode == 0:
            lines = result.stdout.strip().split('\n')
            if len(lines) >= 3:
                # Line 3 format: "((...)) (energy)"
                # We split by space and take the first part
                structure_line = lines[2].strip()
                dot_bracket = structure_line.split(' ')[0]
                return dot_bracket
        print(f"❌ MxFold2 Error: {result.stderr}")
        return "." * len(sequence)

def run_ipknot(sequence):
    """Runs IPKnot with the required LD_LIBRARY_PATH environment variable."""
    # Ensure the environment finds the hidden ViennaRNA libs
    env = os.environ.copy()
    env['LD_LIBRARY_PATH'] = f"{VIENNA_LIB_PATH}:" + env.get('LD_LIBRARY_PATH', '')
    
    with tempfile.NamedTemporaryFile(mode='w', suffix='.fa', delete=True) as tmp:
        tmp.write(f">temp\n{sequence}\n")
        tmp.flush()
        
        cmd = [IPKNOT_BIN, tmp.name]
        result = subprocess.run(cmd, env=env, capture_output=True, text=True)
        
        if result.returncode == 0:
            lines = result.stdout.strip().split('\n')
            if len(lines) >= 3:
                # Line 3 is the structure
                return lines[2].strip()
        print(f"❌ IPKnot Error: {result.stderr}")
        return "." * len(sequence)
```

```python
import numpy as np
import os
from Bio import AlignIO

def generate_msa_covariance(msa_dir, target_id, original_target_id):
    """
    Phase 1: Generates co-evolution signal from a directory of MSAs.
    Constructs path: {msa_dir}/{target_id}.MSA.fasta
    """
    # 1. Construct and Verify Path
    if str(original_target_id) == "__window__":          # sub-sequence folds have no MSA
        return None
    file_path = os.path.join(msa_dir, f"{original_target_id}.MSA.fasta")
    if not os.path.exists(file_path):                    # MSAs are stored per ENTRY (8WT9), not per chain (8WT9_F)
        _base = str(original_target_id).split("_")[0]
        _alt = os.path.join(msa_dir, f"{_base}.MSA.fasta")
        if _base and _alt != file_path and os.path.exists(_alt):
            file_path = _alt

    if not os.path.exists(file_path):
        # FIX: report the ACTUAL path/target we look up (original_target_id, e.g. 8ZNQ),
        # not the per-chain label (8ZNQ_c0). The lookup itself already used the bare target.
        print(f"⚠️ MSA not found for {original_target_id} at {file_path}. Returning empty signal.")
        # We can't determine the sequence length yet, so we return None 
        # to let the consensus function handle it.
        return None

    # 2. Load MSA
    try:
        alignment = AlignIO.read(file_path, "fasta")
    except Exception as e:
        print(f"❌ Error reading MSA for {original_target_id} ({file_path}): {e}")
        return None

    msa_array = np.array([list(rec.seq) for rec in alignment], dtype='U1')
    num_seqs, seq_len = msa_array.shape
    mi_matrix = np.zeros((seq_len, seq_len))

    # 3. Calculate Mutual Information (MI)
    for i in range(seq_len):
        for j in range(i + 1, seq_len):
            cols = msa_array[:, [i, j]]
            valid = (cols[:, 0] != '-') & (cols[:, 1] != '-')
            if np.sum(valid) < (num_seqs * 0.1): 
                continue
                
            pair_data = cols[valid]
            (unique_pairs, counts) = np.unique(pair_data, axis=0, return_counts=True)
            p_xy = counts / len(pair_data)
            
            # Marginal probabilities
            _, counts_x = np.unique(pair_data[:, 0], return_counts=True)
            p_x = counts_x / len(pair_data)
            _, counts_y = np.unique(pair_data[:, 1], return_counts=True)
            p_y = counts_y / len(pair_data)
            
            mi = 0
            # Pre-calculate unique bases for index mapping
            u_x = np.unique(pair_data[:, 0])
            u_y = np.unique(pair_data[:, 1])

            for k, (base_x, base_y) in enumerate(unique_pairs):
                idx_x = np.where(u_x == base_x)[0][0]
                idx_y = np.where(u_y == base_y)[0][0]
                mi += p_xy[k] * np.log(p_xy[k] / (p_x[idx_x] * p_y[idx_y]))
            
            mi_matrix[i, j] = mi_matrix[j, i] = mi

    # 4. Apply APC (Average Product Correction)
    apc_matrix = np.zeros_like(mi_matrix)
    mean_row = np.mean(mi_matrix, axis=1)
    mean_all = np.mean(mi_matrix)
    
    # Avoid division by zero if MI matrix is all zeros
    if mean_all == 0: return mi_matrix

    for i in range(seq_len):
        for j in range(seq_len):
            if i != j:
                apc_matrix[i, j] = mi_matrix[i, j] - (mean_row[i] * mean_row[j] / mean_all)
                
    return apc_matrix
```

```python
def _consensus_ss_uncached(seq, target_id, msa_dir, original_target_id):
    """Phase 2: Consensus 2D Folding with Directory-based MSA lookup."""
    
    # Get 2D predictions
    ss_mxfold = run_mxfold2(seq)
    ss_ipknot = run_ipknot(seq)
    
    # Phase 1: Signal Generation
    cov_signal = generate_msa_covariance(msa_dir, target_id, original_target_id)

    # FIX: the covariance matrix is indexed by MSA-alignment columns; those only
    # line up 1:1 with query residues when the matrix width == len(seq). If it does
    # not, cov_signal[i, partner] mis-indexes or raises IndexError (which would kill
    # the whole run). Disable the signal unless the shapes match.
    if cov_signal is not None and getattr(cov_signal, 'shape', (0,))[0] != len(seq):
        cov_signal = None
    
    consensus = list("." * len(seq))
    
    for i in range(len(seq)):
        m_char = ss_mxfold[i]
        ip_char = ss_ipknot[i]
        
        if m_char == ip_char:
            consensus[i] = m_char
        else:
            # If no MSA signal was found, default to '.' for disagreements
            if cov_signal is None:
                consensus[i] = "."
                continue
                
            # Tie-breaker logic (find_partner helper is required here)
            m_partner = find_partner(ss_mxfold, i)
            ip_partner = find_partner(ss_ipknot, i)
            
            # Extract scores from the matrix
            score_m = cov_signal[i, m_partner] if m_partner != -1 else 0
            score_ip = cov_signal[i, ip_partner] if ip_partner != -1 else 0
            
            if score_m > score_ip and score_m > 0.05:
                consensus[i] = m_char
            elif score_ip > score_m and score_ip > 0.05:
                consensus[i] = ip_char
            else:
                consensus[i] = "."
                
    return "".join(consensus)

def find_partner(dot_bracket, index):
    """
    Finds the index of the base-pairing partner in a dot-bracket string.
    Returns -1 if the base is unpaired ('.').
    """
    if dot_bracket[index] == '.':
        return -1
    
    char = dot_bracket[index]
    # Define pairs and direction
    pairs = { '(': ')', '[': ']', '{': '}', '<': '>',
              ')': '(', ']': '[', '}': '{', '>': '<' }
    
    target = pairs[char]
    step = 1 if char in '([{<' else -1
    
    counter = 0
    for i in range(index, len(dot_bracket) if step == 1 else -1, step):
        if dot_bracket[i] == char:
            counter += 1
        elif dot_bracket[i] == target:
            counter -= 1
            
        if counter == 0:
            return i
            
    return -1
```

## 2D utilities, geometry helpers, consensus-SS cache (new)

```python
import math, json, os, time
import numpy as np

C1_STEP = 5.8   # Å between consecutive C1' atoms in RNA (3.8 Å is the protein CA-CA spacing)

# ── consensus-2D cache (step 2) ──────────────────────────────────────────────
# Requires ONE edit in cell 32: rename `def get_consensus_ss(` to `def _consensus_ss_uncached(`
_SS_CACHE = {}
def get_consensus_ss(seq, target_id, msa_dir, original_target_id):
    key = (seq, original_target_id)
    if key not in _SS_CACHE:
        _SS_CACHE[key] = _consensus_ss_uncached(seq, target_id, msa_dir, original_target_id)
    return _SS_CACHE[key]

# ── pair parsing (all bracket types) + cleaning ──────────────────────────────
_BR = {'(': ')', '[': ']', '{': '}', '<': '>'}
_CL = {v: k for k, v in _BR.items()}
_CANON = {('A','U'), ('U','A'), ('G','C'), ('C','G'), ('G','U'), ('U','G')}

def parse_pairs_all(ss):
    stacks = {o: [] for o in _BR}
    pairs = []
    for i, ch in enumerate(ss):
        if ch in _BR:
            stacks[ch].append(i)
        elif ch in _CL and stacks[_CL[ch]]:
            pairs.append((stacks[_CL[ch]].pop(), i))
    return sorted(pairs)

def clean_pairs(pairs, seq, min_loop=3):
    """Keep canonical, non-hairpin-clash pairs; annotate each with its stem length.
       Returns [(i, j, stem_len)]."""
    keep = {(i, j) for i, j in pairs if j - i > min_loop and (seq[i], seq[j]) in _CANON}
    out = []
    for i, j in sorted(keep):
        run, k = 1, 1
        while (i + k, j - k) in keep: run += 1; k += 1
        k = 1
        while (i - k, j + k) in keep: run += 1; k += 1
        out.append((i, j, run))
    return out

def ss_satisfaction(coords, pairs3, d0=10.5, sigma=2.0):
    """Continuous 2D-agreement in [0,1] (replaces the binary accept/reject gate).
       Base-paired C1'-C1' distance is ~10.5 Å. Stem-weighted. No pairs -> neutral 0.5."""
    if not pairs3: return 0.5
    P = np.asarray(pairs3)
    i, j, run = P[:, 0].astype(int), P[:, 1].astype(int), P[:, 2].astype(float)
    d = np.linalg.norm(coords[i] - coords[j], axis=1)
    s = np.exp(-0.5 * ((d - d0) / sigma) ** 2)
    w = np.minimum(run, 6.0)
    return float((s * w).sum() / w.sum())

def chain_ok(coords, max_step=9.0):
    """Reject models with NaNs or a broken backbone."""
    c = np.asarray(coords, float)
    if c.ndim != 2 or c.shape[1] != 3 or not np.isfinite(c).all(): return False
    if len(c) < 2: return True
    return bool(np.linalg.norm(np.diff(c, axis=0), axis=1).max() <= max_step)

# ── rigid-body helpers ───────────────────────────────────────────────────────
def kabsch_fit(P, Q):
    """R, t such that  P @ R.T + t  ≈ Q  (proper rotation)."""
    cP, cQ = P.mean(0), Q.mean(0)
    U, S, Vt = np.linalg.svd((P - cP).T @ (Q - cQ))
    d = np.sign(np.linalg.det(Vt.T @ U.T)) or 1.0
    R = Vt.T @ np.diag([1.0, 1.0, d]) @ U.T
    return R, cQ - R @ cP

def rmsd_after_kabsch(P, Q):
    R, t = kabsch_fit(P, Q)
    return float(np.sqrt(((P @ R.T + t - Q) ** 2).sum(1).mean()))

def transform_onto(P_all, P_anchor, Q_anchor):
    R, t = kabsch_fit(P_anchor, Q_anchor)
    return P_all @ R.T + t

# ── gap / overhang filling (replaces every "+3.8 Å along +x" heuristic) ──────
def _unit(v):
    n = np.linalg.norm(v)
    return v / n if n > 1e-9 else None

def _solve_arc_half_angle(ratio):
    """Solve sin(x)/x = ratio for x in (0, pi]; ratio = chord / arc length."""
    lo, hi = 1e-6, math.pi
    for _ in range(60):
        mid = 0.5 * (lo + hi)
        if math.sin(mid) / mid > ratio: lo = mid
        else: hi = mid
    return 0.5 * (lo + hi)

def straight_chain(L, step=C1_STEP):
    return np.array([[i * step, 0.0, 0.0] for i in range(L)], dtype=float)

def fill_unmapped(coords, step=C1_STEP):
    """coords: (L,3) with NaN rows for residues without a coordinate.
       Interior gaps  -> circular arc of arc-length (n+1)*step between the two anchors,
                         bulging away from the structure's centre (so it does not clash).
       5'/3' overhangs -> straight tail pointing radially away from the centre."""
    X = np.array(coords, dtype=float, copy=True)
    L = len(X)
    ok = np.isfinite(X).all(1)
    if ok.sum() == 0: return straight_chain(L, step)
    if ok.all(): return X
    idx = np.where(ok)[0]
    cen = X[ok].mean(0)

    def radial(p):
        u = _unit(p - cen)
        return u if u is not None else np.array([1.0, 0.0, 0.0])

    first, last = idx[0], idx[-1]
    d = radial(X[first])
    for k in range(1, first + 1): X[first - k] = X[first] + k * step * d
    d = radial(X[last])
    for k in range(1, L - last): X[last + k] = X[last] + k * step * d

    for a, b in zip(idx[:-1], idx[1:]):
        n = b - a - 1
        if n <= 0: continue
        chord = X[b] - X[a]
        c = np.linalg.norm(chord)
        s = (n + 1) * step
        if c < 1e-6 or c >= 0.9999 * s:
            for k in range(1, n + 1): X[a + k] = X[a] + chord * k / (n + 1)
            continue
        u = chord / c
        mid = 0.5 * (X[a] + X[b])
        r = mid - cen
        perp = _unit(r - np.dot(r, u) * u)
        if perp is None:
            perp = _unit(np.cross(u, [0.0, 0.0, 1.0]))
            if perp is None: perp = _unit(np.cross(u, [0.0, 1.0, 0.0]))
        x = _solve_arc_half_angle(c / s)
        rad = s / (2 * x)
        for k in range(1, n + 1):
            tau = -x + 2 * x * k / (n + 1)
            X[a + k] = mid + rad * math.sin(tau) * u + rad * (math.cos(tau) - math.cos(x)) * perp
    return X

# ── sequence-dependent TM-score (for local validation / threshold calibration) ──
def _tm_d0(L):
    if L < 12: return 0.3
    if L < 16: return 0.4
    if L < 20: return 0.5
    if L < 24: return 0.6
    if L < 30: return 0.7
    return 0.6 * math.sqrt(L - 0.5) - 2.5

def tm_score_fixed(P, Q, n_iter=20):
    """TM-score of prediction P vs reference Q (same residue order; NaN rows ignored)."""
    m = np.isfinite(P).all(1) & np.isfinite(Q).all(1)
    P, Q = P[m], Q[m]
    L = len(Q)
    if L < 3: return 0.0
    d0 = _tm_d0(L)
    d0s = min(max(d0, 4.5), 8.0)
    best = 0.0
    seen = set()
    for frag in sorted({L, max(L // 2, 4), max(L // 4, 4), 4}, reverse=True):
        step = max(1, frag // 2)
        for s0 in range(0, L - frag + 1, step):
            sel = np.arange(s0, s0 + frag)
            for _ in range(n_iter):
                if len(sel) < 3: break
                R, t = kabsch_fit(P[sel], Q[sel])
                d = np.linalg.norm(P @ R.T + t - Q, axis=1)
                best = max(best, float((1.0 / (1.0 + (d / d0) ** 2)).sum() / L))
                cut = d0s
                new = np.where(d < cut)[0]
                while len(new) < 3 and cut < 50:
                    cut += 0.5; new = np.where(d < cut)[0]
                key = tuple(new)
                if key in seen or np.array_equal(new, sel): break
                seen.add(key); sel = new
    return best

# ── statistics collected while the pipeline runs (used by print_tuning_summary / suggest_constants) ──
_STATS = {'geom': [], 'models': []}

def nested_dot_bracket(ss, seq, min_stem=2):
    """Pseudoknot-free, canonical-only, no lone pairs: the kind of input gCoSRNA was trained on
       (junctions parsed from 3D structures). Pairs are kept longest-stem first, crossing pairs dropped."""
    pairs = [p for p in clean_pairs(parse_pairs_all(ss), seq) if p[2] >= min_stem]
    pairs.sort(key=lambda p: (-p[2], p[0]))
    kept = []
    for i, j, r in pairs:
        ok = True
        for k, l, _ in kept:
            if (i < k < j < l) or (k < i < l < j) or len({i, j, k, l}) < 4:   # crossing, or shares a base
                ok = False; break
        if ok: kept.append((i, j, r))
    # a stem that lost partners may now be a lone pair -> drop again
    S = {(i, j) for i, j, _ in kept}
    final = [(i, j) for i, j in S if (i + 1, j - 1) in S or (i - 1, j + 1) in S]
    out = ['.'] * len(seq)
    for i, j in final: out[i], out[j] = '(', ')'
    return ''.join(out)
```

```python
%%writefile /kaggle/working/DRfold2/PotentialFold/operations.py
"""
operations.py: Core Mathematical Operations for RNA Structure Analysis

This module provides essential mathematical operations for manipulating and analyzing
RNA 3D structures, organized into four main categories:

1. Basic Vector Operations:
   Functions for selecting coordinates and calculating distances between points,
   which form the foundation for all structural calculations.

2. Angle Calculations:
   Functions for computing bond angles and dihedral (torsion) angles between atoms,
   with differentiable implementations suitable for gradient-based optimization.

3. Rigid Body Transformations:
   Functions for determining optimal rotations and translations between sets of
   coordinates, enabling structure alignment and manipulation.

4. Sequence Utilities:
   Functions for converting RNA sequence data into standard 3D coordinate templates,
   allowing sequence-structure mapping.

These operations support the core functionality of RNA structure prediction, analysis,
and optimization throughout the codebase.
"""

import os
import torch
import torch.nn as nn
import numpy as np 
import math, sys, math
from io import BytesIO
import torch.nn.functional as F
from torch.autograd import Function
from torch.nn.parameter import Parameter
from subprocess import Popen, PIPE, STDOUT

# Use consistent epsilon value across all functions
EPS = 1e-8


# === Basic Vector Operations ===
def coor_selection(coor,mask):
    #[L,n,3],[L,n],byte
    return torch.masked_select(coor,mask.bool()).view(-1,3)

def pair_distance(x1, x2, eps=1e-6, p=2):
    # Use torch.cdist for p=2 (Euclidean) which is highly optimized
    if p == 2:
        return torch.cdist(x1, x2, p=2)
    
    # For other p-norms, avoid memory expansion with broadcasting
    x1_ = x1.unsqueeze(1)  # [n1, 1, dim]
    x2_ = x2.unsqueeze(0)  # [1, n2, dim]
    diff = torch.abs(x1_ - x2_)
    out = torch.pow(diff + eps, p).sum(dim=2)
    return torch.pow(out, 1. / p)


# === Angle Calculations ===
def angle(p0, p1, p2):
    # [b 3] 
    b0 = p0-p1
    b1 = p2-p1

    b0 = b0 / (torch.norm(b0, dim =-1, keepdim=True) + EPS)
    b1 = b1 / (torch.norm(b1, dim =-1, keepdim=True) + EPS)
    
    recos = torch.sum(b0*b1, -1)
    recos = torch.clamp(recos, -0.9999, 0.9999)
    return torch.acos(recos)

class torsion(Function):
    #PyTorch class to calculate differentiable torsion angle
    #https://stackoverflow.com/questions/20305272/dihedral-torsion-angle-from-four-points-in-cartesian-coordinates-in-python
    #https://salilab.org/modeller/manual/node492.html
    @staticmethod
    def forward(ctx, p0, p1, p2, p3):
        # Save input points for backward pass
        ctx.save_for_backward(p0, p1, p2, p3)

        # Calculate bond vectors
        b0 = p0 - p1
        b1 = p2 - p1
        b2 = p3 - p2

        # Normalize the middle bond vector
        b1_norm = torch.norm(b1, dim=-1, keepdim=True) + 1e-8
        b1_unit = b1 / b1_norm

        # Project the other bonds onto the plane perpendicular to middle bond
        v = b0 - torch.sum(b0 * b1_unit, dim=-1, keepdim=True) * b1_unit
        w = b2 - torch.sum(b2 * b1_unit, dim=-1, keepdim=True) * b1_unit

        # Calculate torsion using the arctan2 formula (more stable than arccos)
        x = torch.sum(v * w, dim=-1)                                # cosine component
        y = torch.sum(torch.cross(b1_unit, v, dim=-1) * w, dim=-1)  # sine component

        return torch.atan2(y, x)

    
    @staticmethod
    def backward(ctx, grad_output):
        # Retrieve saved tensors from forward pass
        p0, p1, p2, p3 = ctx.saved_tensors

        # Calculate bond vectors
        r01 = p0 - p1
        r12 = p2 - p1
        r23 = p3 - p2

        # Calculate bond lengths with numerical stability
        d01 = torch.norm(r01, dim=-1, keepdim=True) + 1e-8
        d12 = torch.norm(r12, dim=-1, keepdim=True) + 1e-8
        d23 = torch.norm(r23, dim=-1, keepdim=True) + 1e-8

        # Normalize bond vectors
        e01 = r01 / d01
        e12 = r12 / d12
        e23 = r23 / d23

        # Calculate normal vectors to the two planes
        n1 = torch.cross(e01, e12, dim=-1)
        n2 = torch.cross(e12, e23, dim=-1)

        # Normalize normal vectors
        n1_norm = torch.norm(n1, dim=-1, keepdim=True) + 1e-8
        n2_norm = torch.norm(n2, dim=-1, keepdim=True) + 1e-8
        n1 = n1 / n1_norm
        n2 = n2 / n2_norm

        # Calculate gradients for each atom
        # These are based on the analytical derivatives of dihedral angles
        g0 = torch.cross(e01, n1, dim=-1) / d01
        g1 = -g0 - torch.cross(e12, n1, dim=-1) / d12
        g2 = torch.cross(e12, n2, dim=-1) / d12 - torch.cross(e23, n2, dim=-1) / d23
        g3 = torch.cross(e23, n2, dim=-1) / d23

        # Apply chain rule with incoming gradient
        g0 = g0 * grad_output.unsqueeze(-1)
        g1 = g1 * grad_output.unsqueeze(-1)
        g2 = g2 * grad_output.unsqueeze(-1)
        g3 = g3 * grad_output.unsqueeze(-1)

        return g0, g1, g2, g3


def dihedral(input1, input2, input3, input4):
    return torsion.apply(input1, input2, input3, input4)



# === Rigid Body Transformations ===
def rigidFrom3Points(x):    
    x1, x2, x3 = x[:, 0], x[:, 1], x[:, 2]
    v1 = x3 - x2
    v2 = x1 - x2
    
    # Normalize v1 to get e1
    e1 = F.normalize(v1, p=2, dim=-1)
    
    # Project v2 onto e1 and subtract to get the component orthogonal to e1
    u2 = v2 - e1 * (torch.einsum('bn,bn->b', e1, v2)[:, None])
    
    # Normalize u2 to get e2
    e2 = F.normalize(u2, p=2, dim=-1)
    
    # Cross product to get e3
    e3 = torch.cross(e1, e2, dim=-1)
    
    return torch.stack([e1, e2, e3], dim=1)


# return the direction from to_q to from_p
def Kabsch_rigid(bases,x1,x2,x3):
    # Early return for empty input
    if x1.shape[0] == 0:
        return torch.empty(0, 3, 3), torch.empty(0, 3)
    
    the_dim=1
    to_q = torch.stack([x1,x2,x3],dim=the_dim)
    biasq=torch.mean(to_q,dim=the_dim,keepdim=True)
    q=to_q-biasq
    m = torch.einsum('bnz,bny->bzy',bases,q)
    u, s, v = torch.svd(m)
    vt = torch.transpose(v, 1, 2)
    det = torch.det(torch.matmul(u, vt))
    det = det.view(-1, 1, 1)
    vt = torch.cat((vt[:, :2, :], vt[:, -1:, :] * det), 1)
    r = torch.matmul(u, vt)
    return r,biasq.squeeze()



# === Sequence Utilities ===
def Get_base(seq,basenpy_standard):
    base_num = basenpy_standard.shape[1]
    basenpy = np.zeros([len(seq),base_num,3])
    seqnpy = np.array(list(seq))
    basenpy[seqnpy=='A']=basenpy_standard[0]
    basenpy[seqnpy=='a']=basenpy_standard[0]

    basenpy[seqnpy=='G']=basenpy_standard[1]
    basenpy[seqnpy=='g']=basenpy_standard[1]

    basenpy[seqnpy=='C']=basenpy_standard[2]
    basenpy[seqnpy=='c']=basenpy_standard[2]

    basenpy[seqnpy=='U']=basenpy_standard[3]
    basenpy[seqnpy=='u']=basenpy_standard[3]

    basenpy[seqnpy=='T']=basenpy_standard[3]
    basenpy[seqnpy=='t']=basenpy_standard[3]
    
    return torch.from_numpy(basenpy).double()
```

```python
%%writefile /kaggle/working/DRfold2/PotentialFold/Optimization.py
#! /nfs/amino-home/liyangum/miniconda3/bin/python
import torch
import random
import numpy as np 
import os, json, sys

import Cubic, Potential
import operations
import a2b, rigid
import torch.optim as opt
from scipy.optimize import minimize
import pickle

torch.manual_seed(6)
np.random.seed(9)
random.seed(9)


Scale_factor = 1.0
USEGEO = False

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")


def readconfig(configfile=''):
    config=[]
    expdir=os.path.dirname(os.path.abspath(__file__))
    if configfile=='':
        configfile=os.path.join(expdir,'lib','ddf.json')
    config=json.load(open(configfile,'r'))
    return config 

    
def inject_pair_prior(dist, pairs, w_base, mu, sigma, lo=2.0, hi=40.0, nbin=36):
    """Mix a base-pair distance prior into a (L,L,nbin+2) distogram of probabilities, IN PLACE.
       bins 1..nbin are the distance bins (2-40 A), bin 0 / bin -1 are the under/over-range bins.
       new = (1-w)*old + w*prior  -> a wrong consensus pair bends the model instead of breaking it.
       Moving mass out of the last (>40 A) bin also lets the pair survive the 'confident' mask."""
    centers = lo + (np.arange(nbin) + 0.5) * (hi - lo) / nbin
    g = np.exp(-0.5 * ((centers - mu) / sigma) ** 2); g = g / g.sum()
    prior = np.zeros(nbin + 2); prior[1:-1] = g
    prior = torch.as_tensor(prior, dtype=dist.dtype, device=dist.device)
    n = 0
    for i, j, run in pairs:
        w = w_base * (1.0 if run >= 3 else (0.5 if run == 2 else 0.0))   # lone pairs are ignored
        if w <= 0: continue
        for a, b in ((i, j), (j, i)):
            dist[a, b, :] = (1.0 - w) * dist[a, b, :] + w * prior
        n += 1
    return n


class Structure:
    def __init__(self,fastafile,geofiles,saveprefix,initial_ret,foldconfig):
        self.config=readconfig(foldconfig)
        self.seqfile=fastafile
        self.init_ret = initial_ret
        self.foldconfig = foldconfig
        self.geofiles = geofiles
        self.rets = [pickle.load(open(refile,'rb')) for refile  in geofiles]
        self.txs=[]
        for ret in self.rets:
            self.txs.append( torch.from_numpy(ret['coor']).double().to(device))
        self.handle_geo()
        self.pair = []
        for ret in self.rets:
            self.pair.append(torch.from_numpy(ret['plddt']).double().to(device))
        self.saveprefix=saveprefix
        self.seq=open(fastafile).readlines()[1].strip()
        self.L=len(self.seq)
        basenpy = np.load(os.path.join(os.path.dirname(os.path.abspath(__file__)),'lib','base.npy'))
        self.basex = operations.Get_base(self.seq,basenpy).to(device)
        othernpy = np.load(os.path.join(os.path.dirname(os.path.abspath(__file__)),'lib','other2.npy'))
        self.otherx = operations.Get_base(self.seq,othernpy).to(device)
        sidenpy = np.load(os.path.join(os.path.dirname(os.path.abspath(__file__)),'lib','side.npy'))
        self.sidex = operations.Get_base(self.seq,sidenpy).to(device)
        
        self.init_mask()
        self.init_paras()
        self._init_fape()
        self.tx2ds = [td.to(device) for td in self.tx2ds]
        self.local_weight = torch.ones(self.L,self.L).to(device)
        
        for i in range(self.L):
            for j in range(i+1,min(self.L,i+2)):
                self.local_weight[i,j] = self.local_weight[j,i] = 4
            for j in range(i+2,min(self.L,i+3)):
                self.local_weight[i,j] = self.local_weight[j,i] = 3
            for j in range(i+3,min(self.L,i+4)):
                self.local_weight[i,j] = self.local_weight[j,i] = 2

    def _init_fape(self):
        self.tx2ds = []
        for tx in self.txs:
            true_rot,true_trans = operations.Kabsch_rigid(self.basex,tx[:,0],tx[:,1],tx[:,2])
            true_x2 = tx[:,None,:,:] - true_trans[None,:,None,:]
            true_x2 = torch.einsum('ijnd,jde->ijne',true_x2,true_rot.transpose(-1,-2))
            self.tx2ds.append(true_x2)
    
    def handle_geo(self):
        oldkeys=['dist_p','dist_c','dist_n']
        newkeys=['pp','cc','nn']
        self.geos=[]
        for ret in self.rets:
            geo = {}
            for nk,ok in zip(newkeys,oldkeys):
                geo[nk] = torch.from_numpy(ret[ok].astype(np.float64)).to(device) + 0
            self.geos.append(geo)
        # ---- 2D restraint: mix a base-pair prior into the C4'-C4' ('cc') and N-N ('nn') distograms ----
        pf = os.environ.get('DRF_PAIRS_FILE', '')
        try:
            pw = float(os.environ.get('DRF_PAIR_W', '0') or 0)
        except ValueError:
            pw = 0.0
        if pf and pw > 0 and os.path.isfile(pf):
            info = json.load(open(pf))
            pairs = [tuple(p) for p in info['pairs']]
            for geo in self.geos:
                if geo['cc'].shape[-1] != 38: continue            # unexpected bin layout -> leave untouched
                n1 = inject_pair_prior(geo['cc'], pairs, pw, info['cc'][0], info['cc'][1])
                n2 = inject_pair_prior(geo['nn'], pairs, pw, info['nn'][0], info['nn'][1])
            print(f"[pair-prior] w={pw} mixed into {n1} cc and {n2} nn pairs", flush=True)
        else:
            print("[pair-prior] OFF for this model", flush=True)


    def init_mask(self):
        halfmask=np.zeros([self.L,self.L])
        fullmask=np.zeros([self.L,self.L])
        for i in range(self.L):
            for j in range(i+1,self.L):
                halfmask[i,j]=1
                fullmask[i,j]=1
                fullmask[j,i]=1
        self.halfmask=(torch.DoubleTensor(halfmask) > 0.5).to(device)
        self.fullmask=(torch.DoubleTensor(fullmask) > 0.5).to(device)
        self.clash_mask = torch.zeros([self.L,self.L,22,22], device=device)
        for i in range(self.L):
            for j in range(i+1,self.L):
                self.clash_mask[i,j]=1

        for i in range(self.L):
             self.clash_mask[i,i,:6,7:]=1

        for i in range(self.L-1):
            self.clash_mask[i,i+1,:,0]=0
            self.clash_mask[i,i+1,0,:]=0
            self.clash_mask[i,i+1,:,5]=0
            self.clash_mask[i,i+1,5,:]=0

        self.side_mask = rigid.side_mask(self.seq).to(device)
        self.side_mask = (self.side_mask[:,None,:,None] * self.side_mask[None,:,None,:]).to(device)
        self.clash_mask = ((self.clash_mask > 0.5) * (self.side_mask > 0.5)).to(device)

        self.geo_confimask_cc = []
        self.geo_confimask_pp = []
        self.geo_confimask_nn = []
        for geo in self.geos:
            confimask_cc = geo['cc'][:,:,-1] < 0.5
            confimask_pp = geo['pp'][:,:,-1] < 0.5
            confimask_nn = geo['nn'][:,:,-1] < 0.5
            self.geo_confimask_cc.append(confimask_cc)
            self.geo_confimask_pp.append(confimask_pp)
            self.geo_confimask_nn.append(confimask_nn)


    def init_paras(self):
        self.geo_cc = []
        self.geo_pp = []
        self.geo_nn = []
        self.cs_coefs = {'cc': [], 'pp': [], 'nn': []}
        self.cs_knots = {'cc': [], 'pp': [], 'nn': []}
        for geo in self.geos:
            cc_cs,cc_decs=Cubic.dis_cubic(geo['cc'],2,40,36)
            pp_cs,pp_decs=Cubic.dis_cubic(geo['pp'],2,40,36)
            nn_cs,nn_decs=Cubic.dis_cubic(geo['nn'],2,40,36)
            self.geo_cc.append([cc_cs,cc_decs])
            self.geo_pp.append([pp_cs,pp_decs])
            self.geo_nn.append([nn_cs,nn_decs])
            
            L = self.L
            cc_coefs_np  = np.stack([[cc_cs[i,j].c for j in range(L)] for i in range(L)], axis=0)
            cc_knots_np  = np.stack([[cc_cs[i,j].x for j in range(L)] for i in range(L)], axis=0)
            self.cs_coefs['cc'].append(torch.from_numpy(cc_coefs_np).to(device))
            self.cs_knots['cc'].append(torch.from_numpy(cc_knots_np).to(device))
            
            pp_coefs_np  = np.stack([[pp_cs[i,j].c for j in range(L)] for i in range(L)], axis=0)
            pp_knots_np  = np.stack([[pp_cs[i,j].x for j in range(L)] for i in range(L)], axis=0)
            self.cs_coefs['pp'].append(torch.from_numpy(pp_coefs_np).to(device))
            self.cs_knots['pp'].append(torch.from_numpy(pp_knots_np).to(device))
            
            nn_coefs_np  = np.stack([[nn_cs[i,j].c for j in range(L)] for i in range(L)], axis=0)
            nn_knots_np  = np.stack([[nn_cs[i,j].x for j in range(L)] for i in range(L)], axis=0)
            self.cs_coefs['nn'].append(torch.from_numpy(nn_coefs_np).to(device))
            self.cs_knots['nn'].append(torch.from_numpy(nn_knots_np).to(device))


    def compute_bb_clash(self,coor,other_coor):
        com_coor = torch.cat([coor,other_coor],dim=1)
        com_dis  = (com_coor[:,None,:,None,:] - com_coor[None,:,None,:,:]).norm(dim=-1)
        dynamicmask2_vdw= (com_dis <= 3.15) * (self.clash_mask)
        vdw_dynamic = Potential.LJpotential(com_dis[dynamicmask2_vdw],3.15)
        return vdw_dynamic.sum()*self.config['weight_vdw']

    def compute_full_clash(self,coor,other_coor,side_coor):
        com_coor = torch.cat([coor[:,:2],other_coor,side_coor],dim=1)
        com_dis  = (com_coor[:,None,:,None,:] - com_coor[None,:,None,:,:]).norm(dim=-1)
        dynamicmask2_vdw= (com_dis <= 2.5) * (self.clash_mask)
        vdw_dynamic = Potential.LJpotential(com_dis[dynamicmask2_vdw],2.5)
        return vdw_dynamic.sum()*self.config['weight_vdw']


    def _cubic_pair_energy(self, atom_map, geo_cs, geo_confimask, weight_key):
        """General cubic-spline energy for CC/PP/NN pairs."""
        min_dis, max_dis, bin_num = 2, 40, 36
        dev = atom_map.device
        upper_th = max_dis - ((max_dis - min_dis) / bin_num) * 0.5
        lower_th = 2.5
        total = torch.zeros((), device=dev, dtype=torch.double)
        spline_key   = weight_key.split('_')[1]  # 'cc', 'pp', or 'nn'
        coeffs_list  = self.cs_coefs[spline_key]
        knots_list   = self.cs_knots[spline_key]
        for block_idx, mask_block in enumerate(geo_confimask):
            mask = (atom_map <= upper_th) & mask_block & self.fullmask & (atom_map >= lower_th)
            idx = mask.nonzero(as_tuple=True)
            if idx[0].numel() > 1:
                coef  = coeffs_list[block_idx][idx]
                knots = knots_list[block_idx][idx]
                part1 = Potential.cubic_distance(atom_map[mask], coef, knots, min_dis, max_dis, bin_num).sum() * self.config[weight_key] * 0.5
            else:
                part1 = torch.zeros((), device=dev)
            part2 = ((atom_map <= lower_th) & mask_block & self.fullmask).sum() * self.config[weight_key]
            total = total + part1 + part2
        return total

    def compute_cc_energy(self, coor):
        atom_map = operations.pair_distance(coor[:,1], coor[:,1])
        return self._cubic_pair_energy(atom_map, self.geo_cc, self.geo_confimask_cc, 'weight_cc')
    
    def compute_pp_energy(self, coor):
        atom_map = operations.pair_distance(coor[:,0], coor[:,0])
        return self._cubic_pair_energy(atom_map, self.geo_pp, self.geo_confimask_pp, 'weight_pp')
    
    def compute_nn_energy(self, coor):
        atom_map = operations.pair_distance(coor[:,-1], coor[:,-1])
        return self._cubic_pair_energy(atom_map, self.geo_nn, self.geo_confimask_nn, 'weight_nn')

    def compute_pccp_energy(self,coor):
        p_atoms=coor[:,0]
        c_atoms=coor[:,1]
        pccpmap=operations.dihedral( p_atoms[self.pccpi], c_atoms[self.pccpi], c_atoms[self.pccpj] ,p_atoms[self.pccpj]                  )
        neg_log = Potential.cubic_torsion(pccpmap,self.pccp_coe,self.pccp_x,36)
        return neg_log.sum()*self.config['weight_pccp']

    def compute_cnnc_energy(self,coor):
        n_atoms=coor[:,-1]
        c_atoms=coor[:,1]
        pccpmap=operations.dihedral( c_atoms[self.cnnci], n_atoms[self.cnnci], n_atoms[self.cnncj] ,c_atoms[self.cnncj]                  )
        neg_log = Potential.cubic_torsion(pccpmap,self.cnnc_coe,self.cnnc_x,36)
        return neg_log.sum()*self.config['weight_cnnc']

    def compute_pnnp_energy(self,coor):
        n_atoms=coor[:,-1]
        p_atoms=coor[:,0]
        pccpmap=operations.dihedral( p_atoms[self.pnnpi], n_atoms[self.pnnpi], n_atoms[self.pnnpj] ,p_atoms[self.pnnpj]                  )
        neg_log = Potential.cubic_torsion(pccpmap,self.pnnp_coe,self.pnnp_x,36)
        return neg_log.sum()*self.config['weight_pnnp']

    def compute_pcc_energy(self,coor):
        p_atoms=coor[:,1]
        c_atoms=coor[:,2]
        pccmap=operations.angle( p_atoms[self.pcci], c_atoms[self.pcci], c_atoms[self.pccj]                   )
        neg_log = Potential.cubic_angle(pccmap,self.pcc_coe,self.pcc_x,12)
        return neg_log.sum()*self.config['weight_pcc']

    def compute_fape_energy(self,coor,ep=1e-3,epmax=20):
        energy= 0
        for tx in self.tx2ds:
            px_mean = coor[:,[1]]
            p_rot   = operations.rigidFrom3Points(coor)
            p_tran  = px_mean[:,0]
            pred_x2 = coor[:,None,:,:] - p_tran[None,:,None,:] # Lx Lrot N , 3
            pred_x2 = torch.einsum('ijnd,jde->ijne',pred_x2,p_rot.transpose(-1,-2)) # transpose should be equal to inverse
            errmap=torch.sqrt( ((pred_x2 - tx)**2).sum(dim=-1) + ep )
            energy = energy + torch.sum(  torch.clamp(errmap,max=epmax)        )
        return energy * self.config['weight_fape']

    def compute_bond_energy(self,coor,other_coor):
        # 3.87
        o3 = other_coor[:-1,-2]
        p  = coor[1:,0]
        dis = (o3-p).norm(dim=-1)
        energy = ((dis-1.607)**2).sum()
        return energy * self.config['weight_bond']

    def tooth_func(self,errmap, ep = 0.05):
        return -1/(errmap/10+ep) + (1/ep)

    def reweight_func(self,ww):
        reweighting = torch.pow(ww,self.config['pair_weight_power'])
        reweighting[ww < self.config['pair_weight_min']] = 0
        return reweighting

    def compute_fape_energy_fromquat(self,x,coor,ep=1e-6,epmax=100):
        energy= 0
        p_rot,px_mean = a2b.Non2rot(x[:,:9],x.shape[0]),x[:,9:]
        pred_x2 = coor[:,None,:,:] - px_mean[None,:,None,:] # Lx Lrot N , 3
        pred_x2 = torch.einsum('ijnd,jde->ijne',pred_x2,p_rot.transpose(-1,-2)) # transpose should be equal to inverse
        for tx,weightplddt in zip(self.tx2ds,self.pair):

            tamplate_dist_map = torch.min( tx.norm(dim=-1), dim=2   )[0]
            errmap=torch.sqrt( ((pred_x2 - tx)**2).sum(dim=-1) + ep ) 
            energy = energy + torch.sum( ( (torch.clamp(errmap,max=self.config['FAPE_max'])**self.config['pair_error_power'])  * self.reweight_func(weightplddt[...,None]) * self.local_weight[...,None] )[tamplate_dist_map>self.config['pair_rest_min_dist']]    )

        return energy * self.config['weight_fape']


    def energy(self,rama):
        coor=a2b.quat2b(self.basex,rama[:,9:])
        other_coor = a2b.quat2b(self.otherx,rama[:,9:])
        side_coor = a2b.quat2b(self.sidex,torch.cat([rama[:,:9],coor[:,-1]],dim=-1))
        
        if self.config['weight_cc']>0:
            E_cc= self.compute_cc_energy(coor) / len(self.rets)
        else:
            E_cc=0
        if self.config['weight_pp']>0:
            E_pp= self.compute_pp_energy(coor) / len(self.rets)
        else:
            E_pp=0
        if self.config['weight_nn']>0:
            E_nn= self.compute_nn_energy(coor) / len(self.rets)
        else:
            E_nn=0

        if self.config['weight_pccp']>0:
            E_pccp= self.compute_pccp_energy(coor) / len(self.rets)
        else:
            E_pccp=0

        if self.config['weight_cnnc']>0:
            E_cnnc= self.compute_cnnc_energy(coor)  / len(self.rets)
        else:
            E_cnnc=0

        if self.config['weight_pnnp']>0:
            E_pnnp= self.compute_pnnp_energy(coor) / len(self.rets)
        else:
            E_pnnp=0

        if self.config['weight_vdw']>0:
            E_vdw= self.compute_full_clash(coor,other_coor,side_coor)
        else:
            E_vdw=0

        if self.config['weight_fape']>0:
            E_fape= self.compute_fape_energy_fromquat(rama[:,9:],coor) / len(self.rets)
        else:
            E_fape=0
        if self.config['weight_bond']>0:
            E_bond= self.compute_bond_energy(coor,other_coor)
        else:
            E_bond=0
        return  E_vdw + E_fape + E_bond + E_pp + E_cc + E_nn + E_pccp + E_cnnc + E_pnnp


    def obj_func_grad_np(self,rama_):
        rama=torch.DoubleTensor(rama_)
        rama.requires_grad=True
        if rama.grad:
            rama.grad.zero_()
        f=self.energy(rama.view(self.L,21))*Scale_factor
        grad_value=autograd.grad(f,rama)[0]
        return grad_value.data.numpy().astype(np.float64)
    
    def obj_func_np(self,rama_):
        rama=torch.DoubleTensor(rama_)
        rama=rama.view(self.L,21)
        with torch.no_grad():
            f=self.energy(rama)*Scale_factor
            return f.item()


    def foldning(self):
        ilter = self.init_ret
        # 1) get initial quaternions (double precision)
        try:
            init_q = self.init_quat(ilter).double()
        except:
            init_q = self.init_quat_safe(ilter).double()

        # 2) move to target device (GPU if available), enable grad
        param = init_q.to(device).clone().detach().requires_grad_(True)

        # 3) set up PyTorch LBFGS optimizer over `param`
        optimizer = opt.LBFGS(
            [param],
            max_iter=self.config.get('max_iter', 300),
            tolerance_grad=1e-6,
            tolerance_change=1e-9,
            history_size=10,
            line_search_fn='strong_wolfe'
        )

        # 4) define the “closure” that LBFGS will call to reevaluate loss + gradients
        def closure():
            optimizer.zero_grad()                                 # clear old grads
            E = self.energy(param.view(self.L,21)) * Scale_factor # compute ∂E/∂param
            E.backward()
            return E

        # 5) run LBFGS until convergence (it calls closure repeatedly)
        optimizer.step(closure)

        # 6) write out final PDB
        final_energy = self.energy(param.view(self.L,21)).item()
        self.outpdb(param, self.saveprefix + '.pdb', energystr=str(final_energy))


    def outpdb(self,rama,savefile,start=0,end=10000,energystr=''):
        # bring baseframes and quaternion data onto CPU to prevent device mismatch
        basex_cpu = self.basex.detach().cpu()
        otherx_cpu = self.otherx.detach().cpu()
        sidex_cpu = self.sidex.detach().cpu()
        shaped_rama = rama.view(self.L,21).detach().cpu()
        # compute backbone and other coords
        coor_np = a2b.quat2b(basex_cpu, shaped_rama[:,9:]).detach().cpu().numpy()
        other_np = a2b.quat2b(otherx_cpu, shaped_rama[:,9:]).detach().cpu().numpy()
        coor = torch.FloatTensor(coor_np)
        # compute side atom coords
        side_coor_NP = a2b.quat2b(sidex_cpu, torch.cat([shaped_rama[:,:9], coor[:,-1]], dim=-1)).detach().cpu().numpy()
        
        Atom_name=[' P  '," C4'",' N1 ']
        Other_Atom_name = [" O5'"," C5'"," C3'"," O3'"," C1'"]
        other_last_name = ['O',"C","C","O","C"]

        side_atoms=         [' N1 ',' C2 ',' O2 ',' N2 ',' N3 ',' N4 ',' C4 ',' O4 ',' C5 ',' C6 ',' O6 ',' N6 ',' N7 ',' N8 ',' N9 ']
        side_last_name =    ['N',      "C",   "O",   "N",   "N",   'N',   'C',   'O',   'C',   'C',   'O',   'N',    'N', 'N','N']

        base_dict = rigid.base_table()
        last_name=['P','C','N']
        wstr=[f'REMARK {str(energystr)}']
        templet='%6s%5d %4s %3s %1s%4d    %8.3f%8.3f%8.3f%6.2f%6.2f          %2s%2s'
        count=1
        for i in range(self.L):
            if self.seq[i] in ['a','g','A','G']:
                Atom_name = [' P  '," C4'",' N9 ']
                #atoms = ['P','C4']

            elif self.seq[i] in ['c','u','C','U']:
                Atom_name = [' P  '," C4'",' N1 ']
            for j in range(coor_np.shape[1]):
                outs=('ATOM  ',count,Atom_name[j],self.seq[i],'A',i+1,coor_np[i][j][0],coor_np[i][j][1],coor_np[i][j][2],0,0,last_name[j],'')
                if i>=start-1 and i < end:
                    wstr.append(templet % outs)
                    count+=1

            for j in range(other_np.shape[1]):
                outs=('ATOM  ',count,Other_Atom_name[j],self.seq[i],'A',i+1,other_np[i][j][0],other_np[i][j][1],other_np[i][j][2],0,0,other_last_name[j],'')
                if i>=start-1 and i < end:
                    wstr.append(templet % outs)
                    count+=1
            
        wstr='\n'.join(wstr)
        wfile=open(savefile,'w')
        wfile.write(wstr)
        wfile.close()
    
    def outpdb_coor(self,coor_np,savefile,start=0,end=1000,energystr=''):
        Atom_name=[' P  '," C4'",' N1 ']
        last_name=['P','C','N']
        wstr=[f'REMARK {str(energystr)}']
        templet='%6s%5d %4s %3s %1s%4d    %8.3f%8.3f%8.3f%6.2f%6.2f          %2s%2s'
        count=1
        for i in range(self.L):
            if self.seq[i] in ['a','g','A','G']:
                Atom_name = [' P  '," C4'",' N9 ']

            elif self.seq[i] in ['c','u','C','U']:
                Atom_name = [' P  '," C4'",' N1 ']
            for j in range(coor_np.shape[1]):
                outs=('ATOM  ',count,Atom_name[j],self.seq[i],'A',i+1,coor_np[i][j][0],coor_np[i][j][1],coor_np[i][j][2],0,0,last_name[j],'')
                if i>=start-1 and i < end:
                    wstr.append(templet % outs)
                count+=1
            
        wstr='\n'.join(wstr)
        wfile=open(savefile,'w')
        wfile.write(wstr)
        wfile.close()


    def init_quat(self,ii):
        x = torch.rand([self.L,21])
        x[:,18:] = self.txs[ii].mean(dim=1)
        init_coor = self.txs[ii]
        biasq = torch.mean(init_coor,dim=1,keepdim=True)
        q = init_coor - biasq
        m = torch.einsum('bnz,bny->bzy',self.basex,q).reshape([self.L,-1])
        x[:,:9] = x[:,9:18] = m
        x.requires_grad_()
        return x

    def init_quat_safe(self,ii):
        x = torch.rand([self.L,21])
        x[:,18:] = self.txs[ii].mean(dim=1)
        init_coor = self.txs[ii]
        biasq = torch.mean(init_coor,dim=1,keepdim=True)
        q = init_coor - biasq + torch.rand([self.L,3,3])
        m = (torch.einsum('bnz,bny->bzy',self.basex,q) + torch.eye(3)[None,:,:]).reshape([self.L,-1])
        x[:,:9] = x[:,9:18] = m
        x.requires_grad_()
        return x


if __name__ == '__main__': 

    fastafile=sys.argv[1]
    saveprefix=sys.argv[2]
    retdirs  =sys.argv[3]
    ret_score = sys.argv[4]
    foldconfig = sys.argv[5]

    savepare = os.path.dirname(saveprefix)
    if not os.path.isdir(savepare):
        os.makedirs(savepare)

    num_of_models = readconfig(foldconfig)['num_of_models']

    score_dict = readconfig(ret_score)
    sorted_items = sorted(score_dict.items(), key=lambda x: x[1])
    lowest_n_keys = [item[0] for item in sorted_items][:num_of_models]
    bestkey = lowest_n_keys[0] + ''
    print("Before sort:", lowest_n_keys)
    lowest_n_keys.sort()
    print("After sort:", lowest_n_keys)
    bestindex = lowest_n_keys.index(bestkey)

    current_ret = bestkey
    retfiles = [os.path.join(retdirs, afile) for afile in lowest_n_keys]
    stru = Structure(fastafile, retfiles, saveprefix + '_from_' + current_ret, bestindex, foldconfig)
    stru.foldning()
```

```python
%%writefile /kaggle/working/DRfold2/PotentialFold/Selection.py
#! /nfs/amino-home/liyangum/miniconda3/bin/python
import numpy
import torch
import torch.autograd as autograd
import numpy as np 

import random
import Cubic, Potential
import operations
import os, json, sys

import a2b, rigid
import torch.optim as opt
from torch.nn.parameter import Parameter
import torch.nn as nn
import math
from scipy.optimize import fmin_l_bfgs_b,fmin_cg,fmin_bfgs
from scipy.optimize import minimize
import lbfgs_rosetta
import pickle
import shutil

torch.manual_seed(6)
torch.set_num_threads(4)
np.random.seed(9)
random.seed(9)

Scale_factor = 1.0
USEGEO = False

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

def readconfig(configfile=''):
    config=[]
    expdir=os.path.dirname(os.path.abspath(__file__))
    if configfile=='':
        configfile=os.path.join(expdir,'lib','ddf.json')
    config=json.load(open(configfile,'r'))
    return config 

    
class Structure:
    def __init__(self, fastafile, geofiles, foldconfig, saveprefix):
        # Load Configuration and Inputs
        self.config = readconfig(foldconfig)
        self.seqfile = fastafile
        self.foldconfig = foldconfig
        self.geofiles = geofiles

        # Load Model Results
        self.rets = [pickle.load(open(refile, 'rb')) for refile  in geofiles]
        
        # Extract Coordinates
        self.txs = []
        for ret in self.rets:
            self.txs.append(torch.from_numpy(ret['coor']).double().to(device))
        
        # Handle Geometrical Data
        self.handle_geo()

        # Extract pLDDT Scores
        self.pair = []
        for ret in self.rets:
            self.pair.append( torch.from_numpy(ret['plddt']).double().to(device))
        
        # Store Output and Sequence Info
        self.saveprefix = saveprefix
        self.seq = open(fastafile).readlines()[1].strip()
        self.L = len(self.seq)
        
        # Load Reference Arrays for Structure Construction
        basenpy = np.load(os.path.join(os.path.dirname(os.path.abspath(__file__)), 'lib', 'base.npy'))
        self.basex = operations.Get_base(self.seq, basenpy).double().to(device)
        
        othernpy = np.load(os.path.join(os.path.dirname(os.path.abspath(__file__)), 'lib', 'other2.npy'))
        self.otherx = operations.Get_base(self.seq, othernpy).double().to(device)
        
        sidenpy = np.load(os.path.join(os.path.dirname(os.path.abspath(__file__)), 'lib', 'side.npy'))
        self.sidex = operations.Get_base(self.seq, sidenpy).double().to(device)        
        
        # Initialize Masks, Parameters, and FAPE
        self.init_mask()
        self.init_paras()
        self._init_fape()
    

    def _init_fape(self):
        self.tx2ds = []
        for tx in self.txs:
            true_rot, true_trans = operations.Kabsch_rigid(self.basex, tx[:, 0], tx[:, 1], tx[:, 2])
            true_x2 = tx[:, None, :, :] - true_trans[None, :, None, :]
            true_x2 = torch.einsum('ijnd,jde->ijne', true_x2, true_rot.transpose(-1,-2))
            self.tx2ds.append(true_x2)
    

    def handle_geo(self):
        oldkeys = ['dist_p', 'dist_c', 'dist_n']
        newkeys = ['pp', 'cc', 'nn']
        self.geos = []
        geo = {'pp':0, 'cc':0, 'nn':0}
        
        for ret in self.rets:    
            for nk, ok in zip(newkeys, oldkeys):
                geo[nk] = geo[nk] + (ret[ok].astype(np.float64) /(len(self.rets)))
        self.geos.append(geo)


    def init_mask(self):
        halfmask=np.zeros([self.L,self.L])
        fullmask=np.zeros([self.L,self.L])
        for i in range(self.L):
            for j in range(i+1,self.L):
                halfmask[i,j]=1
                fullmask[i,j]=1
                fullmask[j,i]=1
        self.halfmask=torch.DoubleTensor(halfmask) > 0.5
        self.fullmask=torch.DoubleTensor(fullmask) > 0.5
        self.clash_mask = torch.zeros([self.L,self.L,22,22])
        for i in range(self.L):
            for j in range(i+1,self.L):
                self.clash_mask[i,j]=1

        for i in range(self.L):
             self.clash_mask[i,i,:6,7:]=1

        for i in range(self.L-1):
            self.clash_mask[i,i+1,:,0]=0
            self.clash_mask[i,i+1,0,:]=0
            self.clash_mask[i,i+1,:,5]=0
            self.clash_mask[i,i+1,5,:]=0

        self.side_mask = rigid.side_mask(self.seq)
        self.side_mask = self.side_mask[:,None,:,None] * self.side_mask[None,:,None,:]
        self.clash_mask = (self.clash_mask > 0.5) * (self.side_mask > 0.5)

        self.geo_confimask_cc = []
        self.geo_confimask_pp = []
        self.geo_confimask_nn = []
        for geo in self.geos:
            confimask_cc = torch.DoubleTensor(geo['cc'][:,:,-1]) < 0.5
            confimask_pp = torch.DoubleTensor(geo['pp'][:,:,-1]) < 0.5
            confimask_nn = torch.DoubleTensor(geo['nn'][:,:,-1]) < 0.5
            self.geo_confimask_cc.append(confimask_cc)
            self.geo_confimask_pp.append(confimask_pp)
            self.geo_confimask_nn.append(confimask_nn)

        # Move masks and confimasks to the GPU/CPU device
        self.halfmask = self.halfmask.to(device)
        self.fullmask = self.fullmask.to(device)
        self.clash_mask = self.clash_mask.to(device)
        self.side_mask = self.side_mask.to(device)
        # geo_confimasks are lists
        self.geo_confimask_cc = [m.to(device) for m in self.geo_confimask_cc]
        self.geo_confimask_pp = [m.to(device) for m in self.geo_confimask_pp]
        self.geo_confimask_nn = [m.to(device) for m in self.geo_confimask_nn]


    def init_paras(self):
        self.geo_cc = []
        self.geo_pp = []
        self.geo_nn = []
        self.cs_coefs = {'cc': [], 'pp': [], 'nn': []}
        self.cs_knots = {'cc': [], 'pp': [], 'nn': []}
        for geo in self.geos:
            cc_cs, cc_decs = Cubic.dis_cubic(geo['cc'], 2, 40, 36)
            pp_cs, pp_decs = Cubic.dis_cubic(geo['pp'], 2, 40, 36)
            nn_cs, nn_decs = Cubic.dis_cubic(geo['nn'], 2, 40, 36)
            self.geo_cc.append([cc_cs, cc_decs])
            self.geo_pp.append([pp_cs, pp_decs])
            self.geo_nn.append([nn_cs, nn_decs])
            L = self.L
            cc_coefs_np = np.stack([[cc_cs[i,j].c for j in range(L)] for i in range(L)], axis=0)
            cc_knots_np = np.stack([[cc_cs[i,j].x for j in range(L)] for i in range(L)], axis=0)
            self.cs_coefs['cc'].append(torch.from_numpy(cc_coefs_np).to(device))
            self.cs_knots['cc'].append(torch.from_numpy(cc_knots_np).to(device))
            pp_coefs_np = np.stack([[pp_cs[i,j].c for j in range(L)] for i in range(L)], axis=0)
            pp_knots_np = np.stack([[pp_cs[i,j].x for j in range(L)] for i in range(L)], axis=0)
            self.cs_coefs['pp'].append(torch.from_numpy(pp_coefs_np).to(device))
            self.cs_knots['pp'].append(torch.from_numpy(pp_knots_np).to(device))
            nn_coefs_np = np.stack([[nn_cs[i,j].c for j in range(L)] for i in range(L)], axis=0)
            nn_knots_np = np.stack([[nn_cs[i,j].x for j in range(L)] for i in range(L)], axis=0)
            self.cs_coefs['nn'].append(torch.from_numpy(nn_coefs_np).to(device))
            self.cs_knots['nn'].append(torch.from_numpy(nn_knots_np).to(device))
     

    def _cubic_pair_energy(self, atom_map, geo_cs, geo_confimask, weight_key):
        """General cubic-spline energy for CC/PP/NN pairs."""
        min_dis, max_dis, bin_num = 2, 40, 36
        dev = atom_map.device
        upper_th = max_dis - ((max_dis - min_dis) / bin_num) * 0.5
        lower_th = 2.5
        total = torch.zeros((), device=dev, dtype=torch.double)
        spline_key = weight_key.split('_')[1]
        coeffs_list = self.cs_coefs[spline_key]
        knots_list = self.cs_knots[spline_key]
        for block_idx, mask_block in enumerate(geo_confimask):
            mask = (atom_map <= upper_th) & mask_block & self.fullmask & (atom_map >= lower_th)
            idx = mask.nonzero(as_tuple=True)
            if idx[0].numel() > 1:
                coef = coeffs_list[block_idx][idx]
                knots = knots_list[block_idx][idx]
                part1 = Potential.cubic_distance(atom_map[mask], coef, knots, min_dis, max_dis, bin_num).sum() * self.config[weight_key] * 0.5
            else:
                part1 = torch.zeros((), device=dev, dtype=torch.double)
            part2 = ((atom_map <= lower_th) & mask_block & self.fullmask).sum() * self.config[weight_key]
            total = total + part1 + part2
        return total

    # GPU-friendly torsion and angle energy helpers
    def _cubic_torsion_energy(self, atom_map, coef, x_vals, weight_key, num_bin):
        energy = Potential.cubic_torsion(atom_map, coef, x_vals, num_bin)
        return energy.sum() * self.config[weight_key]

    def _cubic_angle_energy(self, atom_map, coef, x_vals, weight_key, num_bin):
        energy = Potential.cubic_angle(atom_map, coef, x_vals, num_bin)
        return energy.sum() * self.config[weight_key]

    def compute_cc_energy(self, coor):
        atom_map = operations.pair_distance(coor[:,1], coor[:,1])
        return self._cubic_pair_energy(atom_map, self.geo_cc, self.geo_confimask_cc, 'weight_cc')

    def compute_pp_energy(self, coor):
        atom_map = operations.pair_distance(coor[:,0], coor[:,0])
        return self._cubic_pair_energy(atom_map, self.geo_pp, self.geo_confimask_pp, 'weight_pp')

    def compute_nn_energy(self, coor):
        atom_map = operations.pair_distance(coor[:,-1], coor[:,-1])
        return self._cubic_pair_energy(atom_map, self.geo_nn, self.geo_confimask_nn, 'weight_nn')

    def compute_pccp_energy(self, coor):
        # P-C-C-P dihedral energy on GPU
        p = coor[:, 0]
        c = coor[:, 1]
        dia = operations.dihedral(
            p[self.pccpi], c[self.pccpi], c[self.pccpj], p[self.pccpj]
        )
        return self._cubic_torsion_energy(dia, self.pccp_coe, self.pccp_x, 'weight_pccp', 36)

    def compute_cnnc_energy(self, coor):
        # C-N-N-C dihedral energy on GPU
        n = coor[:, -1]
        c = coor[:, 1]
        dia = operations.dihedral(
            c[self.cnnci], n[self.cnnci], n[self.cnncj], c[self.cnncj]
        )
        return self._cubic_torsion_energy(dia, self.cnnc_coe, self.cnnc_x, 'weight_cnnc', 36)

    def compute_pnnp_energy(self, coor):
        # P-N-N-P dihedral energy on GPU
        n = coor[:, -1]
        p = coor[:, 0]
        dia = operations.dihedral(
            p[self.pnnpi], n[self.pnnpi], n[self.pnnpj], p[self.pnnpj]
        )
        return self._cubic_torsion_energy(dia, self.pnnp_coe, self.pnnp_x, 'weight_pnnp', 36)

    def compute_pcc_energy(self, coor):
        # P-C-C angle energy on GPU
        p = coor[:, 1]
        c = coor[:, 2]
        ang = operations.angle(
            p[self.pcci], c[self.pcci], c[self.pccj]
        )
        return self._cubic_angle_energy(ang, self.pcc_coe, self.pcc_x, 'weight_pcc', 12)

    def compute_fape_energy(self,coor,ep=1e-3,epmax=20):
        energy= 0
        for tx in self.tx2ds:
            px_mean = coor[:,[1]]
            p_rot   = operations.rigidFrom3Points(coor)
            p_tran  = px_mean[:,0]
            pred_x2 = coor[:,None,:,:] - p_tran[None,:,None,:] # Lx Lrot N , 3
            pred_x2 = torch.einsum('ijnd,jde->ijne',pred_x2,p_rot.transpose(-1,-2)) # transpose should be equal to inverse
            errmap=torch.sqrt( ((pred_x2 - tx)**2).sum(dim=-1) + ep )
            energy = energy + torch.sum(  torch.clamp(errmap,max=epmax)        )
        return energy * self.config['weight_fape']

    def compute_bond_energy(self,coor,other_coor):
        # 3.87
        o3 = other_coor[:-1,-2]
        p  = coor[1:,0]
        dis = (o3-p).norm(dim=-1)
        energy = ((dis-1.607)**2).sum()
        return energy * self.config['weight_bond']

    def tooth_func(self,errmap, ep = 0.05):
        return -1/(errmap/10+ep) + (1/ep)
    
    def reweight_func(self,ww):
        reweighting = torch.pow(ww,self.config['pair_weight_power'])
        reweighting[ww < self.config['pair_weight_min']] = 0
        return reweighting
    
    def compute_fape_energy_fromquat(self,x,coor,ep=1e-6,epmax=100):
        energy= 0
        p_rot,px_mean = a2b.Non2rot(x[:,:9],x.shape[0]),x[:,9:]
        pred_x2 = coor[:,None,:,:] - px_mean[None,:,None,:] # Lx Lrot N , 3
        pred_x2 = torch.einsum('ijnd,jde->ijne',pred_x2,p_rot.transpose(-1,-2)) # transpose should be equal to inverse

        for tx,weightplddt in zip(self.tx2ds,self.pair):
            tamplate_dist_map = torch.min( tx.norm(dim=-1), dim=2   )[0]
            errmap=torch.sqrt( ((pred_x2 - tx)**2).sum(dim=-1) + ep ) 
            energy = energy + torch.sum( ( (torch.clamp(errmap,max=self.config['FAPE_max'])**self.config['pair_error_power'])  * self.reweight_func(weightplddt[...,None]) )[tamplate_dist_map>self.config['pair_rest_min_dist']]    )

        return energy * self.config['weight_fape']
    
    def compute_fape_energy_fromcoor(self,coor,ep=1e-6,epmax=100):
        energy= 0
        
        p_rot,px_mean = operations.Kabsch_rigid(self.basex,coor[:,0],coor[:,1],coor[:,2])
        pred_x2 = coor[:,None,:,:] - px_mean[None,:,None,:] # Lx Lrot N , 3
        pred_x2 = torch.einsum('ijnd,jde->ijne',pred_x2,p_rot.transpose(-1,-2)) # transpose should be equal to inverse
        
        for tx,weightplddt in zip(self.tx2ds,self.pair):
            tamplate_dist_map = torch.min( tx.norm(dim=-1), dim=2   )[0]
            errmap=torch.sqrt( ((pred_x2 - tx)**2).sum(dim=-1) + ep ) 
            energy = energy + torch.sum( ( (torch.clamp(errmap,max=self.config['FAPE_max'])**self.config['pair_error_power'])  * self.reweight_func(weightplddt[...,None]) )[tamplate_dist_map>self.config['pair_rest_min_dist']]    )

        return energy * self.config['weight_fape']
    
    
    def energy(self, rama):
        coor = a2b.quat2b(self.basex, rama[:, 9:])
        other_coor = a2b.quat2b(self.otherx, rama[:, 9:])
        side_coor = a2b.quat2b(self.sidex, torch.cat([rama[:, :9], coor[:, -1]], dim=-1))

        E_cc = self.compute_cc_energy(coor) / len(self.geofiles) if self.config['weight_cc'] > 0 else 0
        E_pp = self.compute_pp_energy(coor) / len(self.geofiles) if self.config['weight_pp'] > 0 else 0
        E_nn = self.compute_nn_energy(coor) / len(self.geofiles) if self.config['weight_nn'] > 0 else 0
        E_pccp = self.compute_pccp_energy(coor) / len(self.geofiles) if self.config['weight_pccp'] > 0 else 0
        E_cnnc = self.compute_cnnc_energy(coor) / len(self.geofiles) if self.config['weight_cnnc'] > 0 else 0
        E_pnnp = self.compute_pnnp_energy(coor) / len(self.geofiles) if self.config['weight_pnnp'] > 0 else 0
        E_vdw = self.compute_full_clash(coor, other_coor, side_coor) if self.config['weight_vdw'] > 0 else 0
        E_fape = self.compute_fape_energy_fromquat(rama[:, 9:], coor) / len(self.geofiles) if self.config['weight_fape'] > 0 else 0
        E_bond = self.compute_bond_energy(coor, other_coor) if self.config['weight_bond'] > 0 else 0

        return E_vdw + E_fape + E_bond + E_pp + E_cc + E_nn + E_pccp + E_cnnc + E_pnnp


    def energy_from_coor(self, coor):
        E_cc = self.compute_cc_energy(coor) if self.config['weight_cc'] > 0 else 0
        E_pp = self.compute_pp_energy(coor) if self.config['weight_pp'] > 0 else 0
        E_nn = self.compute_nn_energy(coor) if self.config['weight_nn'] > 0 else 0
        E_fape = (self.compute_fape_energy_fromcoor(coor) / len(self.geofiles)) if self.config['weight_fape'] > 0 else 0
        print(E_fape, E_pp, E_cc, E_nn)
        return E_fape + E_pp + E_cc + E_nn 

    def obj_func_grad_np(self,rama_):
        rama=torch.DoubleTensor(rama_)
        rama.requires_grad=True
        if rama.grad:
            rama.grad.zero_()
        f=self.energy(rama.view(self.L,21))*Scale_factor
        grad_value=autograd.grad(f,rama)[0]
        return grad_value.data.numpy().astype(np.float64)
    
    def obj_func_np(self,rama_):
        rama=torch.DoubleTensor(rama_)
        rama=rama.view(self.L,21)
        with torch.no_grad():
            f = self.energy(rama)*Scale_factor
            return f.item()

    def saveconfig(self,dict,confile):
        json_object = json.dumps(dict, indent = 4)
        wfile = open(confile,'w')
        wfile.write(json_object)
        wfile.close()
    
    def scoring(self):
        geoscale = self.config['geo_scale']
        self.config['weight_pp'] = geoscale * self.config['weight_pp']
        self.config['weight_cc'] = geoscale * self.config['weight_cc']
        self.config['weight_nn'] = geoscale * self.config['weight_nn']
        self.config['weight_pccp'] = geoscale * self.config['weight_pccp']
        self.config['weight_cnnc'] = geoscale * self.config['weight_cnnc']
        self.config['weight_pnnp'] = geoscale * self.config['weight_pnnp']  
        
        energy_dict = {}
        saveenergy_dict  = {}
        
        with torch.no_grad():
            for retfile, tx in zip(self.geofiles, self.txs):
                one = self.energy_from_coor(tx)
                aaretfile = os.path.basename(retfile) 
                energy_dict[aaretfile] = one.item()
                saveenergy_dict[retfile] = one.item()
            self.saveconfig(energy_dict, self.saveprefix)


    def foldning(self):
        minenergy=1e16
        count=0
        for tx in self.txs:
            count+=1
        
        minirama=None

        ilter = self.init_ret
        selected_ret = self.geofiles[ilter]
        try:
            rama=self.init_quat(ilter).data.numpy()
            self.config=readconfig(os.path.join(os.path.dirname(os.path.abspath(__file__)),'lib','vdw.json'))
            rama = fmin_l_bfgs_b(func=self.obj_func_np, x0=rama,  fprime=self.obj_func_grad_np,iprint=10,maxfun=100)[0]
            rama = rama.flatten()
        except:
            rama=self.init_quat_safe(ilter).data.numpy()
            self.config=readconfig(os.path.join(os.path.dirname(os.path.abspath(__file__)),'lib','vdw.json'))
            rama = fmin_l_bfgs_b(func=self.obj_func_np, x0=rama,  fprime=self.obj_func_grad_np,iprint=10,maxfun=100)[0]
            rama = rama.flatten()
            
        self.config=readconfig(self.foldconfig)
        geoscale = self.config['geo_scale']
        self.config['weight_pp'] =geoscale * self.config['weight_pp']
        self.config['weight_cc'] =geoscale * self.config['weight_cc']
        self.config['weight_nn'] =geoscale * self.config['weight_nn']
        self.config['weight_pccp'] =geoscale * self.config['weight_pccp']
        self.config['weight_cnnc'] =geoscale * self.config['weight_cnnc']
        self.config['weight_pnnp'] =geoscale * self.config['weight_pnnp']
        for i in range(3):
            line_min = lbfgs_rosetta.ArmijoLineMinimization(self.obj_func_np,self.obj_func_grad_np,True,len(rama),120)
            lbfgs_opt = lbfgs_rosetta.lbfgs(self.obj_func_np,self.obj_func_grad_np)
            rama=lbfgs_opt.run(rama,256,lbfgs_rosetta.absolute_converge_test,line_min,8000,self.obj_func_np,self.obj_func_grad_np,1e-9)
        newrama=rama+0.0
        newrama=torch.DoubleTensor(newrama) 
        current_energy =self.obj_func_np(rama)

        if current_energy < minenergy:
            print(current_energy,minenergy)
            minenergy=current_energy
            self.outpdb(newrama,self.saveprefix+'.pdb',energystr=str(current_energy))


    def outpdb(self,rama,savefile,start=0,end=10000,energystr=''):
        coor_np=a2b.quat2b(self.basex,rama.view(self.L,21)[:,9:]).data.numpy()
        other_np=a2b.quat2b(self.otherx,rama.view(self.L,21)[:,9:]).data.numpy()
        shaped_rama=rama.view(self.L,21)
        coor = torch.FloatTensor(coor_np)
        side_coor_NP = a2b.quat2b(self.sidex,torch.cat([shaped_rama[:,:9],coor[:,-1]],dim=-1)).data.numpy()
        
        Atom_name=[' P  '," C4'",' N1 ']
        Other_Atom_name = [" O5'"," C5'"," C3'"," O3'"," C1'"]
        other_last_name = ['O',"C","C","O","C"]

        side_atoms=         [' N1 ',' C2 ',' O2 ',' N2 ',' N3 ',' N4 ',' C4 ',' O4 ',' C5 ',' C6 ',' O6 ',' N6 ',' N7 ',' N8 ',' N9 ']
        side_last_name =    ['N',      "C",   "O",   "N",   "N",   'N',   'C',   'O',   'C',   'C',   'O',   'N',    'N', 'N','N']

        base_dict = rigid.base_table()
        
        last_name=['P','C','N']
        wstr=[f'REMARK {str(energystr)}']
        templet='%6s%5d %4s %3s %1s%4d    %8.3f%8.3f%8.3f%6.2f%6.2f          %2s%2s'
        count=1
        for i in range(self.L):
            if self.seq[i] in ['a','g','A','G']:
                Atom_name = [' P  '," C4'",' N9 ']

            elif self.seq[i] in ['c','u','C','U']:
                Atom_name = [' P  '," C4'",' N1 ']
            for j in range(coor_np.shape[1]):
                outs=('ATOM  ',count,Atom_name[j],self.seq[i],'A',i+1,coor_np[i][j][0],coor_np[i][j][1],coor_np[i][j][2],0,0,last_name[j],'')
                if i>=start-1 and i < end:
                    wstr.append(templet % outs)
                    count+=1

            for j in range(other_np.shape[1]):
                outs=('ATOM  ',count,Other_Atom_name[j],self.seq[i],'A',i+1,other_np[i][j][0],other_np[i][j][1],other_np[i][j][2],0,0,other_last_name[j],'')
                if i>=start-1 and i < end:
                    wstr.append(templet % outs)
                    count+=1
            
        wstr='\n'.join(wstr)
        wfile=open(savefile,'w')
        wfile.write(wstr)
        wfile.close()
    
    
    def outpdb_coor(self,coor_np,savefile,start=0,end=1000,energystr=''):
        Atom_name=[' P  '," C4'",' N1 ']
        last_name=['P','C','N']
        wstr=[f'REMARK {str(energystr)}']
        templet='%6s%5d %4s %3s %1s%4d    %8.3f%8.3f%8.3f%6.2f%6.2f          %2s%2s'
        count=1
        for i in range(self.L):
            if self.seq[i] in ['a','g','A','G']:
                Atom_name = [' P  '," C4'",' N9 ']

            elif self.seq[i] in ['c','u','C','U']:
                Atom_name = [' P  '," C4'",' N1 ']
            
            for j in range(coor_np.shape[1]):
                outs=('ATOM  ',count,Atom_name[j],self.seq[i],'A',i+1,coor_np[i][j][0],coor_np[i][j][1],coor_np[i][j][2],0,0,last_name[j],'')
                if i>=start-1 and i < end:
                    wstr.append(templet % outs)
                count+=1
            
        wstr='\n'.join(wstr)
        wfile=open(savefile,'w')
        wfile.write(wstr)
        wfile.close()


if __name__ == '__main__': 

    fastafile = sys.argv[1]
    foldconfig = sys.argv[2]
    save_prefix = sys.argv[3]
    retfiles = sys.argv[4:]

    save_parent_dir = os.path.dirname(save_prefix)
    if not os.path.isdir(save_parent_dir):
        os.makedirs(save_parent_dir)

    retfiles.sort()
    print(retfiles)

    stru = Structure(fastafile, retfiles, foldconfig, save_prefix)    
    stru.scoring()
```

```python
import numpy as np

# ─────────────────────────────────────────────────────────────────────────────
# Coaxial-stacking GEOMETRY
# ─────────────────────────────────────────────────────────────────────────────
# Given a predicted coaxial stack between two stems, rotate the SMALLER ("mobile")
# helix arm so its axis becomes collinear with the other ("reference") helix,
# pivoting at the junction. We move the WHOLE arm the mobile stem subtends
# (range [outer_i .. outer_j]), not just the stem's own paired residues — moving
# only the stem would tear the backbone apart.
#
# ⚠️ This is a HEURISTIC post-processing step. Rigidly re-orienting an arm of an
#    already-predicted 3D model can introduce clashes and may HURT the score as
#    easily as help it. It is therefore: (a) gated by ENABLE_COAXIAL_ALIGNMENT,
#    (b) guarded by a clash check that rejects a rotation bringing the moved arm
#    within COAXIAL_CLASH_THRESHOLD Å of the rest of the structure, and
#    (c) fully wrapped in try/except by the caller. A/B test it against your
#    leaderboard score before trusting it.
ENABLE_COAXIAL_ALIGNMENT = True    # master switch — set False to keep raw 3D coords
COAXIAL_CLASH_THRESHOLD  = 3.0     # Å; reject rotations that clash this tightly
COAXIAL_ANTIPARALLEL     = True    # stacked helices extend opposite ways from junction


def _mid(c, i, j):
    return (c[i] + c[j]) / 2.0

def _stem_axis(c, stem):
    """Helix axis: outer base pair (at the junction) -> innermost base pair."""
    o_i, o_j = stem[0]
    n_i, n_j = stem[-1]
    return _mid(c, n_i, n_j) - _mid(c, o_i, o_j)

def _enclosure(stem):
    """All residue indices the stem subtends (its whole arm)."""
    return set(range(stem[0][0], stem[0][1] + 1))

def _rodrigues(v_from, v_to):
    a = v_from / (np.linalg.norm(v_from) + 1e-9)
    b = v_to   / (np.linalg.norm(v_to)   + 1e-9)
    axis = np.cross(a, b)
    s = np.linalg.norm(axis)
    if s < 1e-8:
        return np.eye(3)
    axis /= s
    ang = np.arccos(np.clip(np.dot(a, b), -1.0, 1.0))
    K = np.array([[0, -axis[2], axis[1]],
                  [axis[2], 0, -axis[0]],
                  [-axis[1], axis[0], 0]])
    return np.eye(3) + np.sin(ang) * K + (1 - np.cos(ang)) * (K @ K)


def align_stems_coaxial(coords, stemA, stemB,
                        antiparallel=COAXIAL_ANTIPARALLEL,
                        clash=COAXIAL_CLASH_THRESHOLD):
    """Make stemA & stemB collinear by rotating the smaller arm about the junction.
       Returns (new_coords, info)."""
    coords = np.asarray(coords, dtype=float)
    eA, eB = _enclosure(stemA), _enclosure(stemB)
    # mobile = smaller arm (the nested child); reference = the larger (parent/longer sibling)
    if len(eA) <= len(eB):
        mob, ref, emob = stemA, stemB, eA
    else:
        mob, ref, emob = stemB, stemA, eB

    a_ref = _stem_axis(coords, ref)
    a_mob = _stem_axis(coords, mob)
    target = -a_ref if antiparallel else a_ref
    R = _rodrigues(a_mob, target)
    pivot = _mid(coords, mob[0][0], mob[0][1])      # mobile arm's junction-proximal base

    new = coords.copy()
    mob_idx = sorted(emob)
    for idx in mob_idx:
        new[idx] = R @ (coords[idx] - pivot) + pivot

    # clash guard: nearest approach between moved atoms and everything that stayed put
    fixed = [i for i in range(len(coords)) if i not in emob]
    if fixed:
        M = new[mob_idx]
        Fx = new[fixed]
        dmin = float(np.sqrt(((M[:, None, :] - Fx[None, :, :]) ** 2).sum(-1)).min())
        if dmin < clash:
            return coords, {"applied": False, "reason": f"min_clash {dmin:.2f}Å < {clash}Å",
                            "mobile_arm": mob_idx}
    else:
        dmin = None
    return new, {"applied": True, "mobile_arm": mob_idx, "min_clash": dmin}


def apply_coaxial_alignment(coords, sequence, coaxial_pairs,
                            antiparallel=COAXIAL_ANTIPARALLEL,
                            clash=COAXIAL_CLASH_THRESHOLD):
    """Apply each predicted coaxial stack to `coords` (in order), with guards."""
    if not ENABLE_COAXIAL_ALIGNMENT or not coaxial_pairs:
        return coords
    coords = np.asarray(coords, dtype=float).copy()
    if coords.shape[0] != len(sequence):
        print(f"   ⚠️ coords/sequence length mismatch "
              f"({coords.shape[0]} vs {len(sequence)}); skipping alignment.")
        return coords
    for pr in coaxial_pairs:
        try:
            coords, info = align_stems_coaxial(coords, pr["stemA"], pr["stemB"],
                                               antiparallel=antiparallel, clash=clash)
            if info["applied"]:
                print(f"   ✅ stacked {pr['label'][0]}↔{pr['label'][1]} "
                      f"(moved {len(info['mobile_arm'])} res, min_clash="
                      f"{info['min_clash']:.2f}Å)" if info['min_clash'] is not None else "")
            else:
                print(f"   ⏭️  skipped {pr['label'][0]}↔{pr['label'][1]} ({info['reason']})")
        except Exception as e:
            print(f"   ⚠️ alignment error on {pr.get('label')}: {e}")
    return coords


# ── kept for backwards-compatibility (the ORIGINAL stem-only rotator) ─────────
# NOTE: this one rotates only the stem's own residues (tears the backbone) and
# took a different signature; it is no longer used by the pipeline. Prefer
# apply_coaxial_alignment() above.
def get_unit_vector(coords, indices):
    """Calculates the normalized direction vector of a stem."""
    vec = coords[indices[-1]] - coords[indices[0]]
    norm = np.linalg.norm(vec)
    return vec / norm if norm > 1e-6 else vec
```

```python
import numpy as np

def kabsch_alignment(P, Q):
    """
    Aligns P (template) to Q (target segment) using the Kabsch algorithm.
    P, Q: NumPy arrays of shape [N, 3].
    Returns the rotated and translated P.
    """
    # 1. Center both sets of coordinates
    centroid_P = np.mean(P, axis=0)
    centroid_Q = np.mean(Q, axis=0)
    P_centered = P - centroid_P
    Q_centered = Q - centroid_Q

    # 2. Computation of the covariance matrix
    H = P_centered.T @ Q_centered

    # 3. Singular Value Decomposition (SVD)
    U, S, Vt = np.linalg.svd(H)
    V = Vt.T
    
    # 4. Correct for reflection (ensure a right-handed coordinate system)
    d = np.linalg.det(V @ U.T)
    step = np.eye(3)
    if d < 0:
        step[2, 2] = -1

    # 5. Optimal Rotation Matrix
    R = V @ step @ U.T

    # 6. Transform P to match Q's pose
    return (R @ P_centered.T).T + centroid_Q

from Bio.Align import PairwiseAligner

# ── graft: scan the training set ONCE per sequence, apply to any number of models ──
_GRAFT_SORTED = None
_GRAFT_CACHE = {}

def _sorted_graft_templates(train_seqs, min_len):
    global _GRAFT_SORTED
    if _GRAFT_SORTED is None or _GRAFT_SORTED[0] is not train_seqs or _GRAFT_SORTED[1] != min_len:
        pairs = [(str(t), str(s)) for t, s in zip(train_seqs['target_id'], train_seqs['sequence'])
                 if len(str(s)) >= min_len]
        pairs.sort(key=lambda p: -len(p[1]))
        _GRAFT_SORTED = (train_seqs, min_len, pairs)
    return _GRAFT_SORTED[2]

def find_graft_matches(sequence, train_seqs, min_len=15):
    key = (sequence, min_len)
    if key in _GRAFT_CACHE: return _GRAFT_CACHE[key]
    matches, regions = [], []
    for t_id, t_seq in _sorted_graft_templates(train_seqs, min_len):
        s = sequence.find(t_seq)
        if s == -1: continue
        e = s + len(t_seq)
        if any(max(s, gs) < min(e, ge) for gs, ge in regions): continue
        matches.append((t_id, s, e)); regions.append((s, e))
    _GRAFT_CACHE[key] = matches
    return matches

def apply_graft_matches(coords, matches, train_coords_dict):
    new = np.array(coords, dtype=float, copy=True)
    for t_id, s, e in matches:
        tc = train_coords_dict.get(t_id)
        if tc is None or len(tc) != e - s or not np.isfinite(tc).all(): continue
        try:
            new[s:e] = kabsch_alignment(tc, new[s:e])
        except Exception as ex:
            print(f"   ⚠️ graft {t_id} failed: {ex}")
    return new

def graft_functional_templates(coords, sequence, train_seqs, train_coords_dict, min_len=15):
    """Drop-in replacement (same signature/behaviour as before, but cached + no per-call sort)."""
    return apply_graft_matches(coords, find_graft_matches(sequence, train_seqs, min_len),
                               train_coords_dict)

# ── template candidates: distinct templates, NaN where unmapped, identity/coverage ──
_TEMPLATE_CACHE = {}

# Same scoring as before, but exposed. Gaps are very cheap here (open -0.5 / extend -0.1 vs match +2), so an
# unrelated sequence can be "aligned" by gapping around mismatches. Candidate confidence below therefore uses
# matches-in-gapless-blocks / max(len(query), len(template)), which punishes gappy alignments. After the LOO calibration you can
# try a stricter aligner, e.g. TPL_ALIGN = (2.0, -1.0, -4.0, -0.5).
TPL_ALIGN = (2.0, -1.0, -0.5, -0.1)    # match, mismatch, gap open, gap extend
TEMPLATE_RANK_BY = "score"             # "score" (alignment score) or "ident" (gapless-block identity)

def _make_aligner():
    al = PairwiseAligner()
    al.match_score, al.mismatch_score, al.open_gap_score, al.extend_gap_score = TPL_ALIGN
    al.mode = 'global'
    return al

def _map_to_query(q_aln, t_aln, t_coords, L, min_block=8):
    """Returns (coords with NaN where unmapped, block_matches, n_aligned).
       block_matches counts identical residues only inside gapless blocks of >= min_block columns: with
       cheap gaps an unrelated sequence still aligns ~65 % of its residues (longest-common-subsequence
       effect), but it essentially never forms long gapless blocks, whereas a real homolog does."""
    X = np.full((L, 3), np.nan)
    qi = ti = aligned = block_matches = 0
    blk_cols = blk_matches = 0
    nt = len(t_coords)
    for qc, tc in zip(q_aln, t_aln):
        if qc != '-' and tc != '-':
            blk_cols += 1; blk_matches += (qc == tc)
            if ti < nt:
                p = t_coords[ti]
                if np.isfinite(p).all() and np.abs(p).max() < 1e5:
                    X[qi] = p; aligned += 1
        else:
            if blk_cols >= min_block: block_matches += blk_matches
            blk_cols = blk_matches = 0
        if qc != '-': qi += 1
        if tc != '-': ti += 1
    if blk_cols >= min_block: block_matches += blk_matches
    return X, block_matches, aligned

def get_template_candidates(sequence, train_seqs, train_coords_dict, k=5, len_tol=0.4,
                            n_align=None, dedupe_rmsd=2.0):
    """Up to k DISTINCT templates, best first. Each: {'t_id','raw'(L,3 with NaN),'ident','cov'}.
       ident = identical residues in gapless blocks (>=8 columns) / max(L, template length) ; cov = aligned residues with coords / L."""
    key = (sequence, k, len_tol)
    if key in _TEMPLATE_CACHE: return _TEMPLATE_CACHE[key]
    L = len(sequence)
    if n_align is None: n_align = 8 if L < 1000 else (4 if L < 3000 else 3)
    al = _make_aligner()
    scored = []
    for t_id, t_seq in zip(train_seqs['target_id'], train_seqs['sequence']):
        t_id, t_seq = str(t_id), str(t_seq)
        if abs(len(t_seq) - L) > len_tol * L: continue
        tc = train_coords_dict.get(t_id)
        if tc is None or len(tc) != len(t_seq): continue
        scored.append((al.score(sequence, t_seq), t_id, t_seq))
    scored.sort(key=lambda x: -x[0])
    out = []
    for _, t_id, t_seq in scored[:max(n_align, k)]:
        aln = al.align(sequence, t_seq)[0]
        X, n_match, n_aln = _map_to_query(str(aln[0]), str(aln[1]), train_coords_dict[t_id], L)
        if not np.isfinite(X).any(): continue
        ident = n_match / max(L, len(t_seq)); cov = n_aln / L
        dup = False
        for o in out:
            m = np.isfinite(X).all(1) & np.isfinite(o['raw']).all(1)
            if m.sum() >= 8 and rmsd_after_kabsch(X[m], o['raw'][m]) < dedupe_rmsd:
                dup = True; break
        if dup: continue
        out.append({'t_id': t_id, 'raw': X, 'ident': ident, 'cov': cov})
        if len(out) >= k: break
    if TEMPLATE_RANK_BY == 'ident': out.sort(key=lambda c: -c['ident'])
    _TEMPLATE_CACHE[key] = out
    return out
```

```python
import os
import subprocess
import tempfile
import numpy as np
import Bio.PDB
import torch
import shutil
from pathlib import Path
from Bio.Align import PairwiseAligner

# --- CONFIGURATION ---
WORKING_DIR = Path("/kaggle/working")
DRFOLD_SCRIPT_DIR = Path("/kaggle/working/DRfold2")

#           with everything below. Keep the helper functions (extract_coords_from_pdb,
#           check_physical_constraints, is_structure_acceptable, parse_pairs) untouched.
# ═════════════════════════════════════════════════════════════════════════════
import threading, glob, signal

PAIR_PRIOR_W        = 0.35        # weight of the base-pair prior mixed into the distogram (0 = off)
PAIR_PRIOR_CC       = (14.83, 0.6) # (mu, sigma) Å, 'cc' distogram of a WC pair (measured: medians 14.8-15.0, within-model IQR 0.1-0.5)
PAIR_PRIOR_NN       = (8.94, 0.6)  # (mu, sigma) Å, 'nn' distogram of a WC pair
BROKEN_PENALTY      = 0.15    # rank penalty for models with a >9 Å backbone step (they still fill slots)
MODEL_RANK_TOL      = 0.05    # agreement buckets: within a bucket DRfold2's own model order is kept
UNRESTRAINED_MODELS = "2,4"       # DRfold2 model indices run WITHOUT the prior (hedge + A/B signal)
PAIR_GEOM_LOG       = True        # print measured C4'-C4' / N-N for confident stems (to calibrate the two lines above)

def extract_c1_strict(pdb_file, expected_len):
    """(L,3) C1' coords or None if the residue count is not exactly expected_len."""
    parser = Bio.PDB.PDBParser(QUIET=True)
    st = parser.get_structure("tmp", str(pdb_file))
    pts = []
    for model in st:
        for chain in model:
            for res in chain:
                key = "C1'" if "C1'" in res else ("C1*" if "C1*" in res else None)
                if key: pts.append(res[key].get_coord())
        break
    return np.array(pts, dtype=float) if len(pts) == expected_len else None

def _measure_pair_geometry(pdb_file, sequence, pairs3):
    """Median and IQR of C4'-C4' and glycosidic N-N over confident (stem>=3) pairs of one model."""
    parser = Bio.PDB.PDBParser(QUIET=True)
    res = list(next(iter(parser.get_structure("g", str(pdb_file)))).get_residues())
    if len(res) != len(sequence): return None
    def atom(r, names):
        for n in names:
            if n in r: return r[n].get_coord()
        return None
    cc, nn = [], []
    for i, j, run in pairs3:
        if run < 3: continue
        a, b = atom(res[i], ["C4'"]), atom(res[j], ["C4'"])
        na = atom(res[i], ["N9"] if sequence[i] in "AG" else ["N1"])
        nb = atom(res[j], ["N9"] if sequence[j] in "AG" else ["N1"])
        if a is not None and b is not None: cc.append(np.linalg.norm(a - b))
        if na is not None and nb is not None: nn.append(np.linalg.norm(na - nb))
    if not cc: return None
    q = lambda v: (float(np.median(v)), float(np.subtract(*np.percentile(v, [75, 25]))))
    return q(cc), (q(nn) if nn else (float('nan'), float('nan'))), len(cc)

def run_drfold2_models(sequence, target_id, original_target_id, pairs3, timeout_s=None):
    """Run DRfold2 once and return ALL relaxed models (up to 5) as a list of (L,3) C1' arrays."""
    fasta_path = WORKING_DIR / f"{target_id}.fasta"
    output_dir = WORKING_DIR / f"preds_{target_id}"
    pairs_path = WORKING_DIR / f"{target_id}.pairs.json"
    with open(fasta_path, "w") as f: f.write(f">{target_id}\n{sequence}\n")

    env = os.environ.copy()
    for k in ("DRF_PAIRS_FILE", "DRF_PAIR_W", "DRF_UNRESTRAINED_MODELS"): env.pop(k, None)
    use_prior = PAIR_PRIOR_W > 0 and any(r >= 2 for _, _, r in pairs3)
    if use_prior:
        with open(pairs_path, "w") as f:
            json.dump({"pairs": [[int(i), int(j), int(r)] for i, j, r in pairs3],
                       "cc": list(PAIR_PRIOR_CC), "nn": list(PAIR_PRIOR_NN)}, f)
        env["DRF_PAIRS_FILE"] = str(pairs_path)
        env["DRF_PAIR_W"] = str(PAIR_PRIOR_W)
        env["DRF_UNRESTRAINED_MODELS"] = UNRESTRAINED_MODELS

    cmd = ["python", "-u", "DRfold_infer.py", str(fasta_path), str(output_dir), "1"]
    proc = subprocess.Popen(cmd, cwd=str(DRFOLD_SCRIPT_DIR), stdout=subprocess.PIPE,
                            stderr=subprocess.STDOUT, text=True, bufsize=1, env=env,
                            start_new_session=True)            # own process group -> we can kill the children too
    def _kill_group():
        try: os.killpg(os.getpgid(proc.pid), signal.SIGKILL)
        except Exception: pass
    timer = None
    if timeout_s:
        timer = threading.Timer(float(timeout_s), _kill_group); timer.daemon = True; timer.start()
    for line in iter(proc.stdout.readline, ''):
        print(f"  [DRfold2 Log]: {line.strip()}", flush=True)
    proc.wait()
    timed_out = timer is not None and not timer.is_alive() and proc.returncode not in (0,)
    if timer is not None: timer.cancel()
    if timed_out: print(f"   ⏱️ DRfold2 killed after {timeout_s:.0f}s for {target_id}")

    models = []
    relax_dir = output_dir / "relax"
    pdbs = sorted(relax_dir.glob("model_*.pdb"), key=lambda p: int(p.stem.split("_")[1]))
    if proc.returncode == 0 and pdbs:
        for p in pdbs:
            c = extract_c1_strict(p, len(sequence))
            if c is not None: models.append(c)
        if PAIR_GEOM_LOG and pairs3:
            _free = {x.strip() for x in UNRESTRAINED_MODELS.split(',') if x.strip()}
            for p in pdbs:
                idx = p.stem.split("_")[1]
                try:
                    g = _measure_pair_geometry(p, sequence, pairs3)
                except Exception as e:
                    g = None
                if g:
                    _STATS['geom'].append({'id': target_id, 'idx': int(idx), 'free': idx in _free,
                                           'cc_med': g[0][0], 'cc_iqr': g[0][1],
                                           'nn_med': g[1][0], 'nn_iqr': g[1][1], 'n': g[2]})
                    print(f"   📏 [pair-geom] {target_id} model_{idx}{'(free)' if idx in _free else ''}: "
                          f"C4'-C4' median={g[0][0]:.2f} IQR={g[0][1]:.2f} | N-N median={g[1][0]:.2f} IQR={g[1][1]:.2f} | n={g[2]}")
        print(f"   🧪 {target_id}: DRfold2 returned {len(models)}/{len(pdbs)} usable models")
    shutil.rmtree(output_dir, ignore_errors=True)
    for p in (fasta_path, pairs_path):
        if p.exists(): os.remove(p)
    return models

_DRF_CACHE = {}
def drfold2_candidates(sequence, chain_id, original_target_id, msa_dir, train_seqs,
                       train_coords_dict, timeout_s=None):
    """DRfold2 (2D-restrained) -> ranked candidate dicts, best 2D agreement first. None on failure.
       The old binary accept/reject gate is gone: broken chains are dropped, everything else is ranked."""
    if sequence in _DRF_CACHE: return _DRF_CACHE[sequence]
    ss = get_consensus_ss(sequence, chain_id, msa_dir, original_target_id)
    pairs3 = clean_pairs(parse_pairs_all(ss), sequence)
    models = run_drfold2_models(sequence, chain_id, original_target_id, pairs3, timeout_s)
    if not models: return None
    matches = find_graft_matches(sequence, train_seqs)
    cands = []
    for k, m in enumerate(models):
        if not np.isfinite(m).all(): continue
        step = float(np.linalg.norm(np.diff(m, axis=0), axis=1).max())          # largest consecutive C1' step
        c = apply_graft_matches(m, matches, train_coords_dict)
        sat = ss_satisfaction(c, pairs3)
        cands.append({'coords': c, 'sat': sat - (BROKEN_PENALTY if step > 9.0 else 0.0), 'raw_sat': sat,
                      'step': step, 'idx': k + 1, 'src': 'drfold',
                      'name': f"drf_m{k+1}{'' if str(k+1) not in UNRESTRAINED_MODELS.split(',') else '(free)'}"})
    if not cands: return None
    cands.sort(key=lambda d: (-round(d['sat'] / MODEL_RANK_TOL), d['idx']))     # saturated agreement -> keep DRfold2's order
    rm = [rmsd_after_kabsch(a['coords'], b['coords']) for i, a in enumerate(cands) for b in cands[i + 1:]]
    print("   🧭 2D agreement / max C1' step per model: " +
          ", ".join(f"{c['name']}={c['raw_sat']:.2f}/{c['step']:.1f}Å" for c in cands))
    if rm: print(f"   🔀 {len(pairs3)} consensus pairs | model-to-model RMSD mean {np.mean(rm):.1f} Å "
                 f"(min {np.min(rm):.1f}, max {np.max(rm):.1f})")
    _STATS['models'].append({'id': chain_id, 'L': len(sequence), 'n_models': len(models), 'n_kept': len(cands),
                             'n_pairs': len(pairs3), 'n_broken': sum(c['step'] > 9.0 for c in cands),
                             'rmsd_mean': float(np.mean(rm)) if rm else None, 'sat': [c['raw_sat'] for c in cands]})
    _DRF_CACHE[sequence] = cands
    return cands

def predict_rna_structures_drfold2(sequence, target_id, original_target_id, msa_dir,
                                   train_seqs, train_coords_dict):
    """Backward-compatible wrapper (legacy cells): best model repeated 5×."""
    c = drfold2_candidates(sequence, target_id, original_target_id, msa_dir, train_seqs, train_coords_dict)
    return None if not c else [c[0]['coords']] * 5

def predict_rna_structures(sequence, target_id, train_seqs, train_coords_dict):
    """Backward-compatible wrapper: up to 5 DISTINCT templates, gaps arc-interpolated at C1' spacing."""
    cands = get_template_candidates(sequence, train_seqs, train_coords_dict, k=5)
    preds = [fill_unmapped(c['raw']) for c in cands]
    if not preds: preds = [straight_chain(len(sequence))]
    while len(preds) < 5: preds.append(preds[0].copy())      # never pad with all-zero models
    return preds

# --- 🛠️ HELPER FUNCTIONS ---

def extract_coords_from_pdb(pdb_file, expected_len):
    parser = Bio.PDB.PDBParser(QUIET=True)
    structure = parser.get_structure("tmp", pdb_file)
    coords = np.zeros((expected_len, 3))
    idx = 0
    for model in structure:
        for chain in model:
            for residue in chain:
                # Handle both prime and asterisk notations
                atom_key = "C1'" if "C1'" in residue else ("C1*" if "C1*" in residue else None)
                if atom_key:
                    coords[idx] = residue[atom_key].get_coord()
                    idx += 1
                if idx >= expected_len: break
        if idx >= expected_len: break
    return coords

def check_physical_constraints(coords, pairs, threshold=11.0):
    # Convert pairs to numpy array for indexing
    pairs = np.array(pairs)
    
    # Extract coordinates for all 'i' and 'j' in one go
    start_coords = coords[pairs[:, 0]]
    end_coords = coords[pairs[:, 1]]
    
    # Calculate all Euclidean distances simultaneously
    dists = np.linalg.norm(start_coords - end_coords, axis=1)
    
    # Calculate metrics
    num_violations = np.sum(dists > threshold)
    max_violation = np.max(dists) if len(dists) > 0 else 0
    accuracy_score = np.sum(dists <= threshold) / len(pairs) if len(pairs) > 0 else 1.0
    
    return {
        "is_valid": num_violations == 0,
        "accuracy_score": accuracy_score, # % of constraints met
        "max_dist": max_violation
    }

def is_structure_acceptable(results, min_accuracy=0.75, extreme_violation_limit=20.0):
    """
    Determines if a DRFold2 structure is physically valid.
    
    Args:
        results (dict): Output from check_physical_constraints_pro
        min_accuracy (float): Minimum % of pairs that must be within threshold (default 85%)
        extreme_violation_limit (float): Absolute max distance allowed for ANY pair (default 20.0A)
        
    Returns:
        bool: True if structure is acceptable, False otherwise.
    """
    # 1. Check if the global topology generally matches the 2D prediction
    has_high_accuracy = results['accuracy_score'] >= min_accuracy
    
    # 2. Check for "Catastrophic Failures" (extreme stretching)
    # Even if accuracy is high, one 50A jump means the chain is broken.
    is_physically_plausible = results['max_dist'] <= extreme_violation_limit
    
    # 3. Robust Decision
    return has_high_accuracy and is_physically_plausible

def parse_pairs(ss):
    stack, pairs = [], []
    for i, char in enumerate(ss):
        if char == '(': stack.append(i)
        elif char == ')':
            if stack: pairs.append((stack.pop(), i))
    return pairs
```

```python
import csv
import numpy as np
import pandas as pd
from collections import defaultdict

# Paths to your Kaggle datasets
seq_path = '/kaggle/input/datasets/jaejohn/rna-all-data/merged_sequences_final.csv'
labels_path = '/kaggle/input/datasets/jaejohn/rna-all-data/merged_labels_final.csv'

# ==========================================
# 1. LOAD SEQUENCE METADATA (train_seqs)
# ==========================================
print("⏳ Loading sequence metadata...")
train_seqs_list = []

with open(seq_path, mode='r', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        # We keep it as a list of dicts first to avoid the read_csv crash
        train_seqs_list.append(row)

# Convert to DataFrame (Usually safer than read_csv in this environment)
train_seqs = pd.DataFrame(train_seqs_list)
print(f"✅ Loaded {len(train_seqs)} sequences.")

# ==========================================
# 2. LOAD COORDINATES (train_coords_dict)
# ==========================================
temp_coords = defaultdict(list)
print("⏳ Pre-indexing training coordinates (Handling missing values)...")

with open(labels_path, mode='r', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        # Validation: Skip rows where coordinates are empty strings
        if not row['x_1'] or not row['y_1'] or not row['z_1']:
            continue 
            
        try:
            # Extract t_id from ID (e.g., 'target_1' -> 'target')
            # This matches your original .rsplit('_', 1)[0] logic
            t_id = row['ID'].rsplit('_', 1)[0]
            
            # Store [resid, x, y, z] for sorting
            temp_coords[t_id].append([
                int(row['resid']), 
                float(row['x_1']), 
                float(row['y_1']), 
                float(row['z_1'])
            ])
        except (ValueError, KeyError):
            # Skip any malformed rows or headers
            continue

# 3. Build one (len(sequence), 3) array per target, NaN where a residue has no coordinate.
#    (The old version dropped missing residues, which silently shifted every later residue when a
#     template was mapped by index.)  resid base is detected (0- or 1-based).
seq_len_map = dict(zip(train_seqs['target_id'], train_seqs['sequence'].str.len()))
_base = min((d[0] for v in temp_coords.values() for d in v), default=1)
print(f"   resid base detected: {_base}")
train_coords_dict, _n_trunc = {}, 0
for t_id, data_list in temp_coords.items():
    n = seq_len_map.get(t_id)
    if n is None: continue
    arr = np.full((n, 3), np.nan)
    for resid, x, y, z in data_list:
        k = resid - _base
        if 0 <= k < n and max(abs(x), abs(y), abs(z)) < 1e5:     # -1e18 style "missing" markers -> NaN
            arr[k] = (x, y, z)
        elif k >= n: _n_trunc += 1
    if np.isfinite(arr).any(): train_coords_dict[t_id] = arr
print(f"   rows with resid beyond sequence length (ignored): {_n_trunc}")

print(f"✅ Successfully indexed {len(train_coords_dict)} structures into train_coords_dict.")
```

```python
def save_coords_to_pdb(sequence, coords, target_id, output_path):
    """Writes C1' coordinates to a PDB file format."""
    with open(output_path, "w") as f:
        f.write(f"HEADER    RNA STRUCTURE PREDICTION - {target_id}\n")
        for i, (char, (x, y, z)) in enumerate(zip(sequence, coords)):
            # PDB ATOM format is very specific about column spacing
            # ATOM  index  name  resname chain resID    x       y       z
            f.write(f"ATOM  {i+1:>5}  C1' {char:>3} A{i+1:>4}    {x:>8.3f}{y:>8.3f}{z:>8.3f}  1.00  0.00           C\n")
        f.write("END\n")
    print(f"💾 Saved: {output_path}")
```

```python
import time
import pandas as pd
import torch
import numpy as np

def run_drfold2_inference(test_csv_path, msa_dir, train_seqs, train_coords_dict, output_csv="submission.csv"):
    """
    Orchestrates the full inference pipeline:
    1. Loads test sequences.
    2. Decides between DRFold2 and Template Matching based on length.
    3. Handles time-outs and GPU memory.
    4. Formats and saves the final submission.
    """
    # Load the target sequences
    test_sequences = pd.read_csv(test_csv_path)
    all_results = []
    start_time = time.time()
    
    print(f"🏁 Starting Inference on {len(test_sequences)} targets...")

    for idx, row in test_sequences.iterrows():
        # 1. 8-hour safety limit (Kaggle limit is 9h, leaving 1h for saving/cleanup)
        elapsed = time.time() - start_time
        if elapsed > (7 * 3600): 
            print("🕒 Time limit approaching (8 hours reached). Wrapping up current results...")
            break 
        
        target_id = str(row['target_id'])
        sequence = str(row['sequence'])
        seq_len = len(sequence)
        
        print(f"\n[{idx+1}/{len(test_sequences)}] 🚀 Processing {target_id} (Length: {seq_len})")

        # 2. Hybrid Decision Logic
        preds = None
        
        if seq_len < 400:
            # Try DRFold2 first for smaller/medium RNAs
            try:
                # FIX: original call omitted `original_target_id`, silently shifting every
                # positional arg. (This run_drfold2_inference is legacy/unused — run_main_loop
                # is the real entry point — but corrected for safety.)
                preds = predict_rna_structures_drfold2(sequence, target_id, target_id, msa_dir, train_seqs, train_coords_dict)
            except Exception as e:
                print(f"⚠️ DRFold2 encountered an error for {target_id}: {e}")
            
            # Fallback to Template Matching if DRFold2 fails or returns None
            if preds is None:
                print(f"🔄 DRFold2 failed or was rejected. Falling back to Template Matching for {target_id}...")
                preds = predict_rna_structures(sequence, target_id, train_seqs, train_coords_dict)
        else:
            # Direct Template Matching for long sequences to avoid OOM
            print(f"📏 Length {seq_len} exceeds DRFold2 threshold. Using Template Matching.")
            preds = predict_rna_structures(sequence, target_id, train_seqs, train_coords_dict)

        # 3. Final Result Verification & Formatting
        # If everything failed, provide zero-coordinates to ensure a valid submission row
        if preds is None or len(preds) < 5:
            print(f"❌ Critical Failure for {target_id}. Filling with zero-coordinates.")
            preds = [[(0.0, 0.0, 0.0)] * seq_len for _ in range(5)]

        # 4. Row-wise Formatting (Competition Schema)
        for j in range(seq_len):
            # The 'ID' is usually target_id_residueIndex
            res_entry = {
                'ID': f"{target_id}_{j+1}", 
                'resname': sequence[j], 
                'resid': j+1
            }
            # Unpack the 5 sets of x, y, z coordinates
            for i in range(5):
                x, y, z = preds[i][j]
                res_entry[f'x_{i+1}'] = x
                res_entry[f'y_{i+1}'] = y
                res_entry[f'z_{i+1}'] = z
            
            all_results.append(res_entry)
        
        # 5. Maintenance
        if torch.cuda.is_available():
            torch.cuda.empty_cache()

    # 6. Final Save
    if not all_results:
        print("Empty results. No submission file generated.")
        return None

    submission_df = pd.DataFrame(all_results)
    submission_df.to_csv(output_csv, index=False)
    
    total_time = (time.time() - start_time) / 3600
    print(f"\n✅ Final Submission Generated: {output_csv}")
    print(f"⏱️ Total runtime: {total_time:.2f} hours")
    
    return submission_df

# --- EXAMPLE CALL ---
# submission = run_drfold2_inference(
#     test_csv_path="/kaggle/input/datasets/khushvaghela/drfold2/test_sequences.csv",
#     msa_dir="/kaggle/input/competition-msas/MSA",
#     train_seqs=train_seqs,
#     train_coords_dict=train_coords_dict
# )
```

```python
import subprocess
import shutil
import os
from pathlib import Path

# --- CONFIGURATION ---
HDOCK_PKG = Path("/kaggle/working/HDOCK_pkg").resolve()

def run_hdock_assembly(receptor_pdb, ligand_pdb, output_folder, target_id):
    """
    Phase 5: Docking with Absolute Path Enforcement.
    """
    # Force everything to absolute paths immediately
    receptor_pdb = Path(receptor_pdb).resolve()
    ligand_pdb = Path(ligand_pdb).resolve()
    output_folder = Path(output_folder).resolve()
    
    # 1. Locate binaries
    hdock_bin = next(HDOCK_PKG.glob("**/hdock*"), None)
    createpl_bin = next(HDOCK_PKG.glob("**/createpl*"), None)
    
    if not hdock_bin or not createpl_bin:
        print("❌ HDOCK binaries not found.")
        return None

    # 2. Setup Intermediate Paths
    work_dir = Path(f"/kaggle/working/dock_{target_id}").resolve()
    work_dir.mkdir(parents=True, exist_ok=True)
    
    dock_out = work_dir / "Hdock.out"
    final_complex_path = output_folder / f"{target_id}_complex_best.pdb"

    print(f"🚢 HDOCK: Docking {target_id}...")

    try:
        # Step A: Run docking using absolute strings
        subprocess.run(
            [str(hdock_bin), str(receptor_pdb), str(ligand_pdb), "-out", str(dock_out)],
            cwd=str(work_dir),
            check=True,
            capture_output=True,
            text=True
        )

        # Step B: Generate top model
        subprocess.run(
            [str(createpl_bin), "Hdock.out", "top1.pdb", "-nmax", "1", "-complex", "-models"],
            cwd=str(work_dir),
            check=True,
            capture_output=True,
            text=True
        )

        model_1 = work_dir / "model_1.pdb"
        if model_1.exists():
            shutil.move(str(model_1), str(final_complex_path))
            shutil.rmtree(work_dir)
            return str(final_complex_path)
            
    except subprocess.CalledProcessError as e:
        print(f"❌ HDOCK failed: {e.stderr}")
        return None
```

```python
import os
import shutil
import subprocess
from pathlib import Path

# --- Global Config ---
TINKER_BIN_DIR = Path("/kaggle/working/tinker_bin")
TINKER_PARAMS = "/kaggle/input/datasets/khushvaghela/tinker-src/tinker-main/tinker-main/params/amber99.prm"

def tinker_refine_pdb(input_pdb, output_dir, gradient=1.0):
    """
    Performs full Tinker refinement using target-specific temporary space.
    """
    # 1. PATH CHANGE: Make work_dir a SUBFOLDER of the target folder
    # This prevents different targets from overwriting each other's temp files
    output_dir = Path(output_dir)
    work_dir = output_dir / "tinker_work" 
    
    if work_dir.exists():
        shutil.rmtree(work_dir)
    work_dir.mkdir(parents=True, exist_ok=True)

    base_name = "refine_task"
    temp_pdb = work_dir / f"{base_name}.pdb"
    final_output_path = output_dir / "final_refined.pdb"

    # 2. PATH CHANGE: Ensure these are defined or passed in!
    # If these are globally defined at the top of your notebook, you are fine.
    # Otherwise, you need to define them here.
    # TINKER_BIN_DIR = Path("/opt/tinker/bin") 
    # TINKER_PARAMS = Path("/opt/tinker/params/amber99.prm")

    # [Sanitization Logic stays the same...]
    rna_map = {'A': 'RA', 'C': 'RC', 'G': 'RG', 'U': 'RU'}
    print(f"🧹 Sanitizing {Path(input_pdb).name} for Tinker...")

    with open(input_pdb, 'r') as f_in, open(temp_pdb, 'w') as f_out:
        for line in f_in:
            if line.startswith("ATOM"):
                res_name = line[17:20].strip()
                if res_name in rna_map:
                    line = line[:17] + rna_map[res_name].ljust(3) + line[20:]
                
                atom_name = line[12:16]
                if "'" in atom_name:
                    line = line[:12] + atom_name.replace("'", "*") + line[16:]
            f_out.write(line)

    # 3. Create Key file inside work_dir
    with open(work_dir / f"{base_name}.key", "w") as k:
        k.write(f"parameters {TINKER_PARAMS}\n")
        k.write("verbose\n")

    # 4. Phase A: pdbxyz (The Mapping)
    # LOGIC TWEAK: pdbxyz sometimes needs more/fewer "1" inputs 
    # depending on the number of chains. "1\n" is usually enough for RNA.
    print("⏳ Phase A: Mapping atoms (pdbxyz)...")
    subprocess.run(
        [str(TINKER_BIN_DIR / "pdbxyz"), base_name],
        cwd=work_dir,
        input="1\n", 
        text=True,
        capture_output=True
    )

    if not (work_dir / f"{base_name}.xyz").exists():
        print(f"❌ Mapping failed for {output_dir.name}. Tinker couldn't recognize residues.")
        return False

    # 5. Phase B: minimize
    print(f"🚀 Phase B: Minimizing (gradient={gradient})...")
    os.environ["OMP_NUM_THREADS"] = "4"
    subprocess.run(
        [str(TINKER_BIN_DIR / "minimize"), base_name, str(gradient)],
        cwd=work_dir,
        input="\n",
        text=True,
        capture_output=True
    )

    # 6. Phase C: xyzpdb (The Export)
    # Tinker versioning: The minimized file is usually base_name.xyz_2
    xyz_out = work_dir / f"{base_name}.xyz_2"
    if not xyz_out.exists():
        xyz_out = work_dir / f"{base_name}.xyz" # Fallback if no movement occurred

    if xyz_out.exists():
        print("⏳ Phase C: Exporting PDB (xyzpdb)...")
        subprocess.run(
            [str(TINKER_BIN_DIR / "xyzpdb"), xyz_out.name],
            cwd=work_dir,
            input="1\n", # Selection for the parameter set
            text=True,
            capture_output=True
        )

        # Tinker output can be .pdb_2 or just .pdb
        pdb_out = work_dir / f"{base_name}.pdb_2"
        if not pdb_out.exists():
            pdb_out = work_dir / f"{base_name}.pdb"

        if pdb_out.exists():
            shutil.move(str(pdb_out), str(final_output_path))
            print(f"🏆 SUCCESS! Saved as: {final_output_path}")
            return True

    print(f"❌ Refinement failed for {output_dir.name}.")
    return False
```

```python
import json
import torch
from protenix.config.config import parse_configs

def build_protenix_configs(input_json_path, dump_dir, model_name="protenix_base_20250630_v1.0.0"):
    # Import config modules from the Protenix code
    from configs.configs_base import configs as configs_base
    from configs.configs_data import data_configs
    from configs.configs_inference import inference_configs
    from configs.configs_model_type import model_configs

    # Merge base configurations
    base = {**configs_base, **{"data": data_configs}, **inference_configs}
    
    # Deep update with model-specific params
    model_spec = model_configs[model_name]
    for k, v in model_spec.items():
        if isinstance(v, dict) and k in base:
            base[k].update(v)
        else:
            base[k] = v

    # CLI-style arguments to override defaults
    arg_str = (
        f"--model_name {model_name} "
        f"--input_json_path {input_json_path} "
        f"--dump_dir {dump_dir} "
        f"--use_msa false "
        f"--use_template false "
        f"--use_rna_msa true "
        f"--sample_diffusion.N_sample 5 "
        f"--seeds 42"
    )
    return parse_configs(configs=base, arg_str=arg_str, fill_required_with_null=True)
```

```python
import re
from Bio.PDB import PDBParser

def extract_all_atoms_from_pdb(pdb_file):
    parser = PDBParser(QUIET=True)
    structure = parser.get_structure("tmp", pdb_file)
    data = []
    for model in structure:
        for chain in model:
            for residue in chain:
                resname = residue.get_resname().strip()
                resid = residue.get_id()[1]
                for atom in residue:
                    coord = atom.get_coord()
                    data.append({'resname': resname, 'resid': resid,
                                 'atom_name': atom.get_name(),
                                 'x': coord[0], 'y': coord[1], 'z': coord[2]})
    return data

# ─────────────────────────────────────────────────────────────────────────────
# FIXED chain parsing
# ─────────────────────────────────────────────────────────────────────────────
# The competition's `sequence` column is the AUTHORITATIVE, scored target: it is
# the concatenation, in `all_sequences` entry order, of every polymer ENTITY
# repeated once per author-chain copy. The number of copies of an entity equals
# the number of `[auth X]` chain IDs listed in that entity's FASTA header — NOT
# the `stoichiometry` column (which the old identify_chains mis-parsed, producing
# the wrong residue count for every multi-copy target and thus an invalid
# submission). We rebuild the chains from the header and then VERIFY the RNA
# concatenation equals `sequence`; the caller falls back to a single-chain model
# if verification ever fails.

def parse_entities(all_sequences_str):
    """Ordered list of unique entities: {'seq', 'n_chains', 'is_rna'}."""
    entities = []
    for entry in str(all_sequences_str).strip().split('>'):
        if not entry.strip():
            continue
        parts = entry.split('|')
        n_chains = 0
        for p in parts:
            if 'Chain' in p:
                auth = re.findall(r'\[auth\s+([A-Za-z0-9]+)\]', p)
                if auth:
                    n_chains = len(auth)
                else:
                    m = re.search(r'Chains?\s+([A-Za-z0-9,\s]+)', p)
                    if m:
                        n_chains = len([c for c in re.split(r'[,\s]+', m.group(1)) if c])
                break
        if n_chains == 0:
            n_chains = 1
        seq = re.sub(r'[^A-Za-z]', '', parts[-1]).upper()
        if not seq:
            continue
        is_rna = all(c in 'ACGUN' for c in seq)
        entities.append({'seq': seq, 'n_chains': n_chains, 'is_rna': is_rna})
    return entities

def build_segments(all_sequences_str, sequence=None):
    """Ordered list of (chain_seq, is_rna), one item per author-chain copy.
       If `sequence` is given and the RNA concatenation does not match it, we
       fall back to a single RNA chain equal to `sequence` (guarantees the
       submission always has exactly len(sequence) residues in the right order)."""
    ents = parse_entities(all_sequences_str)
    segs = []
    for e in ents:
        for _ in range(e['n_chains']):
            segs.append((e['seq'], e['is_rna']))
    if sequence is not None:
        rna_concat = ''.join(s for s, isrna in segs if isrna)
        if rna_concat != str(sequence):
            segs = [(str(sequence), True)]
    return segs

# ── legacy (kept for reference; no longer used by the main loop) ──────────────
def identify_chains(all_sequences_str, stoichiometry_str):
    """DEPRECATED: superseded by build_segments(). The old stoichiometry-based
       copy counting double-counted chains for every multi-copy target."""
    segs = build_segments(all_sequences_str, None)
    return [{'chain': None, 'auth': None,
             'type': 'RNA' if isrna else 'PROTEIN', 'seq': s}
            for s, isrna in segs]
```

```python
def format_submission_row(target_id, refined_atoms):
    formatted_rows = []
    residue_map = {}

    # Standard 3-letter codes for proteins to avoid truncation
    amino_acids = ["ALA", "ARG", "ASN", "ASP", "CYS", "GLN", "GLU", "GLY", "HIS", 
                   "ILE", "LEU", "LYS", "MET", "PHE", "PRO", "SER", "THR", "TRP", "TYR", "VAL"]

    for a in refined_atoms:
        resid = a['resid']
        if resid not in residue_map:
            residue_map[resid] = {
                'resname': a['resname'],
                'coord': None
            }

        # ✅ FIX: Look for C1' (Nucleic) OR CA (Protein)
        atom_name = a['atom_name'].strip()
        if atom_name in ["C1'", "C1*", "C1", "CA"]:
            residue_map[resid]['coord'] = (a['x'], a['y'], a['z'])

    for resid in sorted(residue_map.keys()):
        res = residue_map[resid]
        resname = res['resname'].strip()
        coord = res['coord']

        if coord is None:
            # Re-enable the warning to help debug missing chain segments
            print(f"⚠️ Warning: No representative atom (C1'/CA) found for {target_id} residue {resid}.")
            continue 

        # ✅ FIX: Normalization that respects Proteins
        if resname in ["A", "C", "G", "U"]:
            resname_sub = resname
        elif resname.upper() in amino_acids:
            resname_sub = resname # Keep full 3-letter code for proteins
        elif resname.startswith("R"): # RA, RC...
            resname_sub = resname[1]
        elif resname.startswith("D"): # DNA
            resname_sub = resname
        else:
            resname_sub = resname[-1] # Fallback for single-letter formats

        row_entry = {
            'ID': f"{target_id}_{resid}",
            'resname': resname_sub,
            'resid': resid
        }

        # Repeat coordinates for all 5 sample columns
        for i in range(1, 6):
            row_entry[f'x_{i}'], row_entry[f'y_{i}'], row_entry[f'z_{i}'] = coord

        formatted_rows.append(row_entry)

    print(f"📊 Formatted {len(formatted_rows)} residues (RNA + Protein) for {target_id}")
    return formatted_rows
```

```python
import math, time
import numpy as np

# ---- time budget (step 3) ---------------------------------------------------
NOTEBOOK_LIMIT_S = 8 * 3600     # hard limit of the notebook run
SAFETY_S         = 20 * 60      # kept free for CSV writing / slow last target
SETUP_OFFSET_S   = 600          # seconds already burned before run_main_loop starts (your log: ~456 s)
PLAN_FRACTION    = 0.90         # only plan to use 90 % of the remaining time
DOCK_S           = 90.0         # rough HDOCK cost per docking step (update after you see a log)
CHAIN_MAX_LEN    = 300          # longest single chain DRfold2 may fold (its spline fit is O(L^2) and 5x repeated)

# ---- template policy ---------------------------------------------------------
TEMPLATE_SKIP_CONF  = 0.90   # best-template identity (matches/L) at/above which DRfold2 is skipped  ← tune by LOO
WHOLE_TEMPLATE_CONF = 0.50   # multi-chain targets: use the template's own chain arrangement above this
HIGH_TEMPLATE_ID    = 0.80   # slot order: templates first above this
MID_TEMPLATE_ID     = 0.50   # slot order: DRfold2 first, template second above this

# ---- long-target tiers (step 5) ---------------------------------------------
DUP_TM              = 0.90   # two candidates with mutual TM-score >= this are the same hypothesis (chains <= DUP_TM_MAX_L)
DUP_TM_MAX_L        = 400
DUP_RMSD            = 1.0    # fallback duplicate test (Å RMSD) for chains longer than DUP_TM_MAX_L
MIN_TEMPLATE_CONF   = 0.15   # extra templates below this identity are dropped (the best one is always kept)
LONG_SEG_FOLD_SLOTS = 2      # only the first N template slots get DRfold2-filled gaps
SEG_MIN, SEG_MAX_WIN, SEG_FLANK = 25, 150, 8

class TimeModel:
    """t(L) = a + b*L^2 fitted by least squares on observed per-chain DRfold2 wall times
       (target-start -> done, incl. SS, graft, coaxial). Seeded with the 4 runs in your log
       and refitted online after every real run."""
    def __init__(self, prior):
        self.obs = list(prior)
    def add(self, L, secs):
        self.obs.append((float(L), float(secs)))
    def coef(self):
        X = np.array([[1.0, L * L] for L, _ in self.obs]); y = np.array([t for _, t in self.obs])
        (a, b), *_ = np.linalg.lstsq(X, y, rcond=None)
        return max(a, 60.0), max(b, 0.0)
    def predict(self, L):
        a, b = self.coef()
        return a + b * L * L

TM = TimeModel([(30, 213.0), (59, 283.0), (69, 333.0), (101, 419.0), (165, 1155.0),
                (214, 1761.0), (246, 2484.0), (255, 2622.0), (267, 3029.0),
                (19, 205.0), (28, 163.0), (34, 209.0), (59, 295.0), (71, 350.0), (101, 449.0)])   # measured wall times
_RUN_T0 = [time.time()]
_SLACK_S = [0.0]

def time_left():
    used = (time.time() - NB_T0) if 'NB_T0' in globals() else SETUP_OFFSET_S + (time.time() - _RUN_T0[0])
    return NOTEBOOK_LIMIT_S - SAFETY_S - used

def _template_cost(L):          # template mapping cost; ~130 s for the 4640-nt target in your log
    return 5.0 + 6e-6 * L * L

def _template_confident(chain_seq, train_seqs, train_coords_dict):
    c = get_template_candidates(chain_seq, train_seqs, train_coords_dict, k=5)
    return bool(c) and c[0]['ident'] >= TEMPLATE_SKIP_CONF

def _plan_drfold(infos, tl):
    """Greedy cheapest-first admission under the remaining time. Returns (set(target ids), slack seconds)."""
    reserve = sum(i['fallback'] for i in infos)
    budget = PLAN_FRACTION * tl - reserve
    chosen, spent = set(), 0.0
    for i in sorted((i for i in infos if i['schedulable']), key=lambda i: _target_cost(i)):
        c = _target_cost(i)
        if spent + c <= budget:
            chosen.add(i['tid']); spent += c
    return chosen, max(0.0, budget - spent)

def _target_cost(info):
    n_dock = (len(info['rna']) - 1) if (ENABLE_COMPLEX_DOCKING and len(info['rna']) > 1) else 0
    return 1.15 * sum(TM.predict(len(s)) for s in info['need']) + DOCK_S * n_dock

def _slack_allows(L):
    return L <= CHAIN_MAX_LEN and 1.15 * TM.predict(L) <= 0.5 * _SLACK_S[0]

# ---- DRfold2 on a window (used by the long-target tiers) --------------------
def fold_window(subseq, wid, msa_dir, train_seqs, train_coords_dict):
    if subseq in _DRF_CACHE: return _DRF_CACHE[subseq]
    if not _slack_allows(len(subseq)): return None
    t0 = time.time()
    try:
        cands = drfold2_candidates(subseq, wid, "__window__", msa_dir, train_seqs, train_coords_dict,
                                   timeout_s=max(900.0, 3 * TM.predict(len(subseq))))
    except Exception as e:
        print(f"     ⚠️ window fold failed ({wid}): {e}"); return None
    dt = time.time() - t0
    if cands:
        TM.add(len(subseq), dt); _SLACK_S[0] = max(0.0, _SLACK_S[0] - dt)
    return cands

def _nan_runs(mask):
    runs, s = [], None
    for i, m in enumerate(mask):
        if m and s is None: s = i
        if not m and s is not None: runs.append((s, i)); s = None
    if s is not None: runs.append((s, len(mask)))
    return runs

def complete_with_segments(raw, sequence, chain_id, slot_k, msa_dir, train_seqs, train_coords_dict):
    """Tier B/C: residues the template does not cover (NaN runs >= SEG_MIN) are folded by
       2D-restrained DRfold2 in windows <= SEG_MAX_WIN (+SEG_FLANK context each side) and stitched
       by Kabsch-aligning the flank residues that already have coordinates. Windows are processed
       left->right so a placed window becomes the anchor for the next. Anything left over is
       arc-interpolated later by fill_unmapped."""
    X = np.array(raw, dtype=float, copy=True)
    L = len(X)
    for (s, e) in _nan_runs(~np.isfinite(X).all(1)):
        if e - s < SEG_MIN: continue
        n_win = math.ceil((e - s) / SEG_MAX_WIN); size = math.ceil((e - s) / n_win)
        for w in range(n_win):
            ws, we = s + w * size, min(e, s + (w + 1) * size)
            a, b = max(0, ws - SEG_FLANK), min(L, we + SEG_FLANK)
            cands = fold_window(sequence[a:b], f"{chain_id}_w{a}", msa_dir, train_seqs, train_coords_dict)
            if not cands: continue
            M = cands[slot_k % len(cands)]['coords']
            anchors = np.array([i for i in range(a, b) if not (ws <= i < we) and np.isfinite(X[i]).all()], int)
            if len(anchors) >= 3:  M2 = transform_onto(M, M[anchors - a], X[anchors])
            elif len(anchors) > 0: M2 = M + (X[anchors[0]] - M[anchors[0] - a])
            else:                  M2 = M
            X[ws:we] = M2[ws - a:we - a]
            print(f"     🧵 {chain_id}: folded+stitched window {ws+1}-{we} ({len(anchors)} anchors)")
    return X

def template_models(sequence, chain_id, msa_dir, train_seqs, train_coords_dict,
                    k=5, allow_seg_fold=False, len_tol=0.4):
    """Candidate dicts from DISTINCT templates, best first (long-target tiers):
         A  good template        -> one template per slot, gaps arc-filled at C1' spacing
         B  partial coverage     -> uncovered runs folded by DRfold2 + stitched (first slots, if time allows)
         C  no template at all   -> windows folded by DRfold2 + stitched (if time allows) else straight chain
       Exact-sequence motifs from the training set are grafted onto every candidate."""
    L = len(sequence)
    cands = get_template_candidates(sequence, train_seqs, train_coords_dict, k=k, len_tol=len_tol)
    print(f"     📚 {chain_id}: {len(cands)} distinct template(s), identity {[round(c['ident'], 2) for c in cands]}, "
          f"coverage {[round(c['cov'], 2) for c in cands]}")
    matches = find_graft_matches(sequence, train_seqs)
    jobs = [(c['raw'], c['ident'], c['cov'], c['t_id']) for c in cands]
    jobs = jobs[:1] + [j for j in jobs[1:] if j[1] >= MIN_TEMPLATE_CONF]
    if not jobs:
        n = LONG_SEG_FOLD_SLOTS if allow_seg_fold else 1
        jobs = [(np.full((L, 3), np.nan), 0.0, 0.0, 'none')] * n
    out = []
    for slot_k, (raw, ident, cov, t_id) in enumerate(jobs):
        X = raw
        if allow_seg_fold and slot_k < LONG_SEG_FOLD_SLOTS:
            X = complete_with_segments(X, sequence, chain_id, slot_k, msa_dir, train_seqs, train_coords_dict)
        X = fill_unmapped(X)
        X = apply_graft_matches(X, matches, train_coords_dict)
        out.append({'coords': X, 'conf': ident, 'cov': cov, 'src': 'template', 'name': f"tpl_{t_id}"})
    return out

# ---- slot diversity (step 1) -------------------------------------------------
def build_slots(drf, tmpl, n=5, dup_rmsd=None):
    """Fill the n submission slots with DIFFERENT hypotheses (best-of-5 TM scoring rewards that).
       drf: DRfold2 candidates (best 2D agreement first); tmpl: template candidates (best identity first).
         strong template (id>=HIGH)  : T D T D D
         medium template (id>=MID)   : D T D T D
         weak / none                 : D D D D T"""
    if dup_rmsd is None: dup_rmsd = DUP_RMSD
    D, T = list(drf), list(tmpl)
    top_t = T[0]['conf'] if T else 0.0
    pattern = "TDTDD" if top_t >= HIGH_TEMPLATE_ID else ("DTDTD" if top_t >= MID_TEMPLATE_ID else "DDDDT")
    chosen = []
    def is_dup(c):
        a = c['coords']
        if len(a) < 3: return False
        for o in chosen:
            if len(a) <= DUP_TM_MAX_L:
                if tm_score_fixed(a, o['coords']) >= DUP_TM: return True     # TM-score is what the competition measures
            elif rmsd_after_kabsch(a, o['coords']) < dup_rmsd: return True
        return False
    def take(src):
        while src:
            c = src.pop(0)
            if not is_dup(c): return c
        return None
    for tag in pattern:
        first, second = (T, D) if tag == 'T' else (D, T)
        c = take(first) or take(second)
        if c is not None: chosen.append(c)
    for c in D + T:                              # leftovers if some slots are still empty
        if len(chosen) >= n: break
        if not is_dup(c): chosen.append(c)
    if not chosen: return []
    while len(chosen) < n: chosen.append(chosen[len(chosen) % len(chosen)])
    return chosen[:n]
```

```python
# ─────────────────────────────────────────────────────────────────────────────
# SUBMISSION-AUTHORITATIVE ORCHESTRATION
# ─────────────────────────────────────────────────────────────────────────────
# The submission MUST contain exactly len(sequence) rows per target, IDs
# {target_id}_{1..N}, resname = sequence[i]. So we build the output directly from
# the `sequence` column: predict one 3D monomer per unique RNA chain, then lay the
# chains into a full-length coordinate array in the exact order that reconstructs
# `sequence`. Any failure degrades to a valid placeholder chain.
#
# HOW SCORING SHAPES THIS DESIGN (Stanford RNA 3D Folding 2):
#   • Scored with TM-score on C1' atoms only; the prediction is superimposed onto
#     the reference by US-align (a global rotation+translation, sequence-independent),
#     best-of-5, averaged over targets.
#   • Because US-align applies ONE global transform to the whole prediction, a rigid
#     move of the entire structure does not change the score. Consequently:
#       - For a SINGLE RNA chain, only its internal conformation matters.
#       - For MULTIPLE RNA chains (homo/hetero multimers: 9G4P, 9G4Q, 9WHV, 9I9W,
#         9JGM, 9MME, 9ZCC), the RELATIVE placement of the chains matters, since one
#         global transform must fit them all — this is where docking helps.
#       - A non-scored DNA/protein partner does NOT change the RNA's internal fold
#         (HDOCK is rigid-body; DRfold2 folds the RNA monomer independently), so
#         docking a partner is score-NEUTRAL for the RNA C1' residues. It can still
#         be used as a scaffold if you want (ENABLE_PARTNER_PROTENIX below).
#
# DRFOLD_MAX_LEN is gone. DRfold2 admission is now decided by the time-budget scheduler
# (see the "SCHEDULER" cell: CHAIN_MAX_LEN, TimeModel, _plan_drfold) on UNIQUE chain sequences.

INTER_CHAIN_OFFSET = 60.0    # Å translation between assembled chains (avoid overlap)

# --- assembly strategy for MULTI-CHAIN targets ---
# False -> fast, always-valid: place each chain's monomer with a fixed offset
#          (correct internal folds; relative placement is only a guess).
# True  -> dock chains into one complex with HDOCK to get real relative placement,
#          then extract ONLY the RNA C1' atoms (in order) for the rows. Slower and
#          depends on the HDOCK binaries; falls back to offset assembly on any
#          failure or residue-count mismatch. (Requires setup_hdock() to have run.)
ENABLE_COMPLEX_DOCKING = True
# When docking, also predict DNA/protein partner chains with Protenix and include
# them in the docking scaffold. Score-neutral for the RNA rows (see note above);
# only affects the docking geometry. Needs a loaded `shared_runner`.
ENABLE_PARTNER_PROTENIX = False

_MONOMER_CACHE = {}          # cache monomer coords by chain sequence

def _straight_chain(L):
    return straight_chain(L)          # C1' spacing (5.8 Å), not the 3.8 Å protein CA spacing

def _to_coords_array(coords, L):
    """Coerce anything to a clean (L,3) float array; short/NaN tails are filled by fill_unmapped."""
    try:
        a = np.asarray(coords, dtype=float)
    except Exception:
        a = np.empty((0, 3))
    if a.ndim != 2 or a.shape[1] != 3 or a.shape[0] == 0:
        return _straight_chain(L)
    a = a[:L]
    if a.shape[0] < L:
        a = np.vstack([a, np.full((L - a.shape[0], 3), np.nan)])
    return fill_unmapped(a) if not np.isfinite(a).all() else a

def predict_monomer_candidates(chain_seq, chain_id, original_target_id, msa_dir,
                               train_seqs, train_coords_dict, use_drfold2):
    """Return 5 slot-ordered (L,3) arrays for ONE RNA chain (cached per sequence).
       DRfold2 (2D-restrained, all 5 relaxed models kept) + distinct templates, mixed by build_slots.
       DRfold2 is skipped when a template is already confident, or when the scheduler says no."""
    key = (chain_seq, bool(use_drfold2))
    if key in _MONOMER_CACHE:
        return [c.copy() for c in _MONOMER_CACHE[key]]
    L = len(chain_seq)
    confident = _template_confident(chain_seq, train_seqs, train_coords_dict)
    run_drf = bool(use_drfold2) and L <= CHAIN_MAX_LEN and not confident
    if use_drfold2 and confident:
        print(f"     ⏭️  {chain_id}: confident template (identity ≥ {TEMPLATE_SKIP_CONF}) -> skipping DRfold2")

    drf = []
    if run_drf:
        t0 = time.time()
        try:
            drf = drfold2_candidates(chain_seq, chain_id, original_target_id, msa_dir,
                                     train_seqs, train_coords_dict,
                                     timeout_s=min(max(900.0, 3 * TM.predict(L)), max(120.0, time_left()))) or []
            if drf: TM.add(L, time.time() - t0)
        except Exception as e:
            print(f"     ⚠️ DRfold2 failed for {chain_id}: {e}")
    try:
        tmpl = template_models(chain_seq, chain_id, msa_dir, train_seqs, train_coords_dict,
                               k=5, allow_seg_fold=not run_drf)
    except Exception as e:
        print(f"     ⚠️ Template matching failed for {chain_id}: {e}"); tmpl = []

    slots = build_slots(drf, tmpl)
    if not slots:
        print(f"     ↪️  {chain_id}: using straight-chain placeholder ({L} nt).")
        coords_list = [_straight_chain(L) for _ in range(5)]
    else:
        print("     🎰 slots: " + " | ".join(s['name'] for s in slots))
        coords_list = [_to_coords_array(s['coords'], L) for s in slots]

    if L <= CHAIN_MAX_LEN:                       # coaxial stacking on slot 1 only (raw copy kept as a hedge)
        try:
            ss = nested_dot_bracket(get_consensus_ss(chain_seq, chain_id, msa_dir, original_target_id), chain_seq)
            cx = get_coaxial_stem_pairs(chain_seq, ss)
            if cx:
                raw0 = coords_list[0]
                adj = _to_coords_array(apply_coaxial_alignment(raw0, chain_seq, cx), L)
                if not np.allclose(adj, raw0, atol=1e-3):
                    print(f"     📐 {chain_id}: applying {len(cx)} coaxial stack(s) to slot 1.")
                    coords_list[0] = adj
                    coords_list[-1] = raw0.copy()
        except Exception as e:
            print(f"     ⚠️ Coaxial refinement skipped for {chain_id}: {e}")

    _MONOMER_CACHE[key] = [c.copy() for c in coords_list]
    return coords_list

def predict_monomer_coords(chain_seq, chain_id, original_target_id, msa_dir,
                           train_seqs, train_coords_dict, use_drfold2):
    """Slot-1 coordinates of one chain (kept so _assemble_offset/_assemble_docked work unchanged)."""
    return predict_monomer_candidates(chain_seq, chain_id, original_target_id, msa_dir,
                                      train_seqs, train_coords_dict, use_drfold2)[0]

# ── OFFSET assembly (default): fast + always valid ───────────────────────────
def _assemble_offset(rna_chains, target_id, msa_dir, train_seqs, train_coords_dict,
                     use_drfold2):
    parts = []
    for idx, cseq in enumerate(rna_chains):
        mc = predict_monomer_coords(cseq, f"{target_id}_c{idx}", target_id,
                                    msa_dir, train_seqs, train_coords_dict, use_drfold2)
        parts.append(mc + np.array([idx * INTER_CHAIN_OFFSET, 0.0, 0.0]))
    return np.vstack(parts) if parts else None

# ── DOCKED assembly (optional): real relative placement via HDOCK ─────────────
_NUC = {"A", "C", "G", "U", "RA", "RC", "RG", "RU", "N"}
def _is_rna_res(resname):
    r = resname.strip().upper()
    return r in _NUC or (len(r) == 1 and r in "ACGU") or (r.startswith("R") and r[1:] in {"A","C","G","U"})

def _extract_rna_c1_in_order(pdb_path):
    """Return C1' coords of every RNA residue in the complex, in file order."""
    atoms = extract_all_atoms_from_pdb(str(pdb_path))
    out, seen = [], set()
    for a in atoms:
        if not _is_rna_res(a['resname']):
            continue
        an = a['atom_name'].strip()
        if an in ("C1'", "C1*", "C1"):
            out.append((a['x'], a['y'], a['z']))
    return np.array(out, dtype=float) if out else np.empty((0, 3))

def _raw_c1_from_pdb(pdb_path):
    """C1' atoms read straight from the ATOM/HETATM records, in file order (no residue-name filter, no chain merging)."""
    pts, tags = [], []
    for line in open(pdb_path):
        if line.startswith(("ATOM", "HETATM")) and line[12:16].strip() in ("C1'", "C1*"):
            pts.append((float(line[30:38]), float(line[38:46]), float(line[46:54])))
            tags.append((line[21], line[17:20].strip()))
    return (np.array(pts, dtype=float) if pts else np.empty((0, 3))), tags

def _assemble_docked(segs, target_id, msa_dir, train_seqs, train_coords_dict,
                     shared_runner, expected_len, use_drfold2):
    """Predict every chain, dock them in sequence order into one complex, then
       pull back ONLY the RNA C1' atoms (in order). Falls back (returns None) on
       any failure or if the RNA residue count != expected_len."""
    try:
        target_dir = Path(target_id); target_dir.mkdir(exist_ok=True)
        chain_pdbs = []
        for idx, (cseq, is_rna) in enumerate(segs):
            cid = f"{target_id}_c{idx}"
            pdb_path = target_dir / f"{cid}.pdb"
            if is_rna:
                mc = predict_monomer_coords(cseq, cid, target_id, msa_dir,
                                            train_seqs, train_coords_dict, use_drfold2)
                save_coords_to_pdb(cseq, mc, cid, str(pdb_path))
                chain_pdbs.append(str(pdb_path))
            elif ENABLE_PARTNER_PROTENIX and shared_runner is not None:
                mol = "DNA" if ("T" in cseq and "U" not in cseq) else "PROTEIN"
                run_protenix_fast(sequence=cseq, target_id=cid, mol_type=mol,
                                  output_pdb_path=str(pdb_path), runner=shared_runner)
                if pdb_path.exists():
                    chain_pdbs.append(str(pdb_path))
            # else: partner chain skipped (score-neutral for RNA rows)

        if not chain_pdbs:
            return None

        current = chain_pdbs[0]
        for j in range(1, len(chain_pdbs)):
            docked = run_hdock_assembly(receptor_pdb=current, ligand_pdb=chain_pdbs[j],
                                        output_folder=target_dir,
                                        target_id=f"{target_id}_dock_{j}")
            if docked:
                current = docked
            else:
                print(f"   ⚠️ HDOCK step {j} failed; falling back to offset assembly.")
                return None

        coords = _extract_rna_c1_in_order(current)
        if coords.shape[0] != expected_len:
            raw, tags = _raw_c1_from_pdb(current)
            from collections import Counter
            print(f"   🔎 docked PDB C1' atoms by (chain, resname): {dict(Counter(tags))}")
            if raw.shape[0] == expected_len:
                print("   ✅ raw C1' parse matches the expected length; using it.")
                coords = raw
        if coords.shape[0] != expected_len:
            print(f"   ⚠️ docked RNA residue count {coords.shape[0]} != {expected_len}; "
                  f"falling back to offset assembly.")
            return None
        return coords
    except Exception as e:
        print(f"   ⚠️ docked assembly error ({e}); falling back to offset assembly.")
        return None

def assemble_full_coords(segs, target_id, msa_dir, train_seqs, train_coords_dict,
                         shared_runner=None, use_drfold2=True):
    """Produce (len(sequence),3) coords. Uses docking if enabled (with fallback),
       otherwise the fast offset assembly. Returns None only if no RNA chains.

       `use_drfold2` reflects the per-target decision (based on OVERALL sequence
       length) made once in run_main_loop, and is forwarded to every chain."""
    rna_chains = [s for s, is_rna in segs if is_rna]
    if not rna_chains:
        return None
    expected_len = sum(len(s) for s in rna_chains)

    if ENABLE_COMPLEX_DOCKING and len(rna_chains) > 1:
        docked = _assemble_docked(segs, target_id, msa_dir, train_seqs,
                                  train_coords_dict, shared_runner, expected_len,
                                  use_drfold2)
        if docked is not None:
            return docked
    return _assemble_offset(rna_chains, target_id, msa_dir, train_seqs,
                            train_coords_dict, use_drfold2)

def rows_for_slots(target_id, sequence, slots):
    """Emit exactly len(sequence) rows; model column k holds slots[k-1] (5 different hypotheses)."""
    L = len(sequence)
    slots = [_to_coords_array(s, L) for s in slots]
    while len(slots) < 5: slots.append(slots[0].copy())
    rows = []
    for i, ch in enumerate(sequence):
        entry = {'ID': f"{target_id}_{i+1}", 'resname': ch, 'resid': i + 1}
        for k in range(1, 6):
            x, y, z = slots[k - 1][i]
            entry[f'x_{k}'], entry[f'y_{k}'], entry[f'z_{k}'] = float(x), float(y), float(z)
        rows.append(entry)
    return rows

def rows_for_target(target_id, sequence, coords):
    return rows_for_slots(target_id, sequence, [coords] * 5)

def _whole_template_slots(target_id, sequence, msa_dir, train_seqs, train_coords_dict):
    """'whole' route: the template's own chain arrangement is kept (no per-chain split, no docking).
       Up to 5 DISTINCT templates, one per slot; uncovered runs folded by DRfold2 when time allows."""
    L = len(sequence)
    tm = template_models(sequence, target_id, msa_dir, train_seqs, train_coords_dict,
                         k=5, allow_seg_fold=True, len_tol=0.5)
    slots = build_slots([], tm)
    if not slots: return None
    print("     🎰 slots: " + " | ".join(s['name'] for s in slots))
    return [_to_coords_array(s['coords'], L) for s in slots]

def assemble_full_slots(segs, target_id, msa_dir, train_seqs, train_coords_dict,
                        shared_runner=None, use_drfold2=True):
    """5 full-length arrays. Slot 1 = the normal (docked/offset) assembly. For multi-chain targets the
       other slots reuse slot 1's chain arrangement: each chain's k-th candidate is Kabsch-fitted onto the
       corresponding chain of slot 1, so there is no extra HDOCK run."""
    rna = [s for s, is_rna in segs if is_rna]
    if not rna: return None
    base = assemble_full_coords(segs, target_id, msa_dir, train_seqs, train_coords_dict,
                                shared_runner, use_drfold2=use_drfold2)
    if base is None: return None
    cands = {s: _MONOMER_CACHE[(s, bool(use_drfold2))] for s in set(rna)}
    if len(rna) == 1:
        return [base] + [c.copy() for c in cands[rna[0]][1:]]
    slots = [base]
    for k in range(1, 5):
        parts, off = [], 0
        for cs in rna:
            ref = base[off:off + len(cs)]
            c = cands[cs][k]
            parts.append(transform_onto(c, c, ref) if len(cs) >= 4 else c)
            off += len(cs)
        slots.append(np.vstack(parts))
    return slots

def analyze_target(row, msa_dir, train_seqs, train_coords_dict):
    """Cheap per-target analysis used by the scheduler (template lookups are cached and reused later)."""
    seq, L = str(row.sequence), len(str(row.sequence))
    info = {'tid': str(row.target_id), 'seq': seq, 'L': L, 'route': 'whole', 'segs': None,
            'rna': [], 'need': [], 'schedulable': False, 'fallback': _template_cost(L)}
    try:
        segs = build_segments(row.all_sequences, seq)
    except Exception as e:
        print(f"   ⚠️ build_segments failed for {info['tid']} ({e}); whole-structure route.")
        return info
    rna = [s for s, is_rna in segs if is_rna]
    if not rna: return info
    uniq = list(dict.fromkeys(rna))
    if all(len(s) <= CHAIN_MAX_LEN for s in uniq):
        info.update(route='chain', segs=segs, rna=rna)
        if len(rna) > 1:
            wc = get_template_candidates(seq, train_seqs, train_coords_dict, k=5, len_tol=0.5)
            if wc and wc[0]['ident'] >= WHOLE_TEMPLATE_CONF:
                info['route'] = 'whole'
        if info['route'] == 'chain':
            info['need'] = [s for s in uniq if not _template_confident(s, train_seqs, train_coords_dict)]
            info['schedulable'] = bool(info['need'])
    return info

def run_main_loop(input_csv_path, msa_dir, train_seqs, train_coords_dict,
                  shared_runner=None, output_csv="submission.csv", select=None):
    """Every target yields a schema-correct block of rows; a failing target degrades to a placeholder."""
    df = pd.read_csv(input_csv_path)
    all_rows, walls = [], []
    start = time.time(); _RUN_T0[0] = start
    print(f"🏁 Starting inference on {len(df)} targets...  time left: {time_left()/3600:.2f} h")

    infos = [analyze_target(r, msa_dir, train_seqs, train_coords_dict) for r in df.itertuples()]
    if select is not None:                      # TEST MODE: run only a stratified subset
        keep = select(infos)
        mask = [inf['tid'] in keep for inf in infos]
        df = df[mask].reset_index(drop=True); infos = [inf for inf, m in zip(infos, mask) if m]
    print("📋 plan (route / schedulable / est. DRfold2 cost):")
    for i in infos:
        print(f"   {i['tid']:>8}  L={i['L']:<5} {i['route']:<5} "
              f"{'DRfold-eligible ~%4.0fs' % _target_cost(i) if i['schedulable'] else '-'}")

    for pos, (row, info) in enumerate(zip(df.itertuples(), infos), 1):
        target_id, sequence, L = info['tid'], info['seq'], info['L']
        t_target = time.time()
        tl = time_left()
        chosen, slack = _plan_drfold(infos[pos - 1:], tl)
        _SLACK_S[0] = slack
        use_dr = target_id in chosen
        label = ('whole-structure template' if info['route'] == 'whole'
                 else ('chain: DRfold2+templates' if use_dr else 'chain: templates only'))
        print(f"\n[{pos}/{len(df)}] 🚀 {target_id} (len {L}) [{label}]  "
              f"time left {tl/3600:.2f} h, slack {slack:.0f}s")

        if tl <= 0:
            print("🕒 Time budget reached; filling with placeholders.")
            all_rows.extend(rows_for_slots(target_id, sequence, [_straight_chain(L)] * 5))
            continue
        try:
            if info['route'] == 'whole':
                slots = _whole_template_slots(target_id, sequence, msa_dir, train_seqs, train_coords_dict)
            else:
                slots = assemble_full_slots(info['segs'], target_id, msa_dir, train_seqs,
                                            train_coords_dict, shared_runner, use_drfold2=use_dr)
            if slots is None or any(len(s) != L for s in slots):
                print(f"   ⚠️ assembly length mismatch; using placeholder.")
                slots = [_straight_chain(L)] * 5
        except Exception as e:
            print(f"   ❌ {target_id} failed ({e}); using placeholder.")
            slots = [_straight_chain(L)] * 5

        all_rows.extend(rows_for_slots(target_id, sequence, slots))
        walls.append((target_id, L, use_dr, time.time() - t_target))
        print(f"   ✅ {target_id}: wrote {L} residues in {time.time()-t_target:.0f}s.")
        if torch.cuda.is_available():
            torch.cuda.empty_cache()

    cols = ['ID', 'resname', 'resid', 'x_1', 'y_1', 'z_1', 'x_2', 'y_2', 'z_2',
            'x_3', 'y_3', 'z_3', 'x_4', 'y_4', 'z_4', 'x_5', 'y_5', 'z_5']
    submission_df = pd.DataFrame(all_rows)[cols]
    submission_df.to_csv(output_csv, index=False)
    print(f"\n📦 Wrote {len(submission_df)} rows for "
          f"{submission_df['ID'].str.rsplit('_', n=1).str[0].nunique()} targets -> {output_csv}")
    a, b = TM.coef()
    print(f"⏱️ Runtime: {(time.time()-start)/3600:.2f} h   | fitted t(L) = {a:.0f} + {b:.5f}·L²  "
          f"(paste these into TimeModel's prior next run)")
    print("   per-target wall times:", [(t, L, 'DR' if d else '--', round(w)) for t, L, d, w in walls])
    return submission_df
```

## Tuning tools (stats summary, test subset, validation harnesses)

```python
# ═════════════════════════════════════════════════════════════════════════════
# NEW CELL "TUNING TOOLS"  (after the leave-one-out cell, before the final run cell)
# ═════════════════════════════════════════════════════════════════════════════
import json, time
import numpy as np
import pandas as pd

# ───────── 1. suggestions computed from the statistics the pipeline collects while it runs ─────────
def suggest_constants(verbose=True):
    out = {}
    G = [g for g in _STATS['geom'] if g['n'] >= 5 and g['nn_med'] < 10.5 and g['cc_med'] < 17]   # well-formed stems only
    free = [g for g in G if g['free']]
    use = free if len(free) >= 5 else G
    if len(use) >= 5:
        for key, m, q in (('PAIR_PRIOR_CC', 'cc_med', 'cc_iqr'), ('PAIR_PRIOR_NN', 'nn_med', 'nn_iqr')):
            mu = float(np.median([g[m] for g in use]))
            sg = float(np.clip(np.median([g[q] for g in use]) / 1.35, 0.6, 1.5))
            out[key] = (round(mu, 2), round(sg, 2))
        if verbose: print(f"   geometry from {len(use)} {'free' if use is free else 'all'}-model measurements")
    return out

def print_tuning_summary():
    M = _STATS['models']
    print("\n══════════ TUNING SUMMARY ══════════")
    if M:
        rm = [m['rmsd_mean'] for m in M if m['rmsd_mean'] is not None]
        print(f"DRfold2 runs: {len(M)} | models kept per run: {np.mean([m['n_kept'] for m in M]):.1f}/5 | "
              f"broken-backbone models: {sum(m['n_broken'] for m in M)}/{sum(m['n_kept'] for m in M)}")
        if rm: print(f"model-to-model RMSD per run: median {np.median(rm):.1f} Å (min {np.min(rm):.1f}, max {np.max(rm):.1f})")
        zero = [m['id'] for m in M if m['sat'] and max(m['sat']) < 0.05 and m['n_pairs'] > 0]
        nopair = [m['id'] for m in M if m['n_pairs'] == 0]
        print(f"targets with NO consensus pairs: {nopair}")
        print(f"targets where every model disagrees with the consensus (agreement≈0): {zero}")
    a, b = TM.coef()
    print(f"time model now: t(L) = {a:.0f} + {b:.5f}·L²   (paste into TimeModel prior if you like)")
    s = suggest_constants()
    print("suggested constants from this run:", s if s else "(not enough measurements yet)")

# ───────── 2. small, stratified TEST subset (RUN_MODE="test") ─────────
def select_test_subset(infos, n_short=5, max_short_len=130, n_multi=1, n_tpl=1, n_long=1):
    """Covers every code path with few targets: short single-chain DRfold2 targets spread over the length range,
       one multi-chain target (docking + slots 2-5), one template-confident target (DRfold2 skipped),
       one whole-structure/long target (template tiers)."""
    keep = []
    short = sorted([i for i in infos if i['route'] == 'chain' and i['schedulable'] and len(i['rna']) == 1
                    and i['L'] <= max_short_len], key=lambda i: i['L'])
    if short:
        idx = np.unique(np.linspace(0, len(short) - 1, min(n_short, len(short))).round().astype(int))
        keep += [short[k]['tid'] for k in idx]
    multi = sorted([i for i in infos if i['route'] == 'chain' and len(i['rna']) > 1], key=_target_cost)
    keep += [i['tid'] for i in multi[:n_multi]]
    tpl = sorted([i for i in infos if i['route'] == 'chain' and not i['schedulable']], key=lambda i: i['L'])
    keep += [i['tid'] for i in tpl[:n_tpl]]
    long_ = sorted([i for i in infos if i['route'] == 'whole'], key=lambda i: i['L'])
    keep += [i['tid'] for i in long_[:n_long]]
    keep = list(dict.fromkeys(keep))
    est = sum(_target_cost(i) for i in infos if i['tid'] in keep and i['schedulable'])
    print(f"🧪 test subset: {keep}  (est. DRfold2 time ≈ {est/60:.0f} min)")
    return set(keep)

# ───────── 3. validation harnesses ─────────
def _val_pool(train_seqs, train_coords_dict, min_len, max_len):
    return [(str(t), str(s)) for t, s in zip(train_seqs['target_id'], train_seqs['sequence'])
            if min_len <= len(str(s)) <= max_len and str(t) in train_coords_dict
            and np.isfinite(train_coords_dict[str(t)]).all(1).mean() > 0.9]

def pick_validation_set(train_seqs, train_coords_dict, n=6, min_len=40, max_len=90, max_ident=0.5, seed=1):
    """Training RNAs WITHOUT a close template in the rest of the training set (like the hard test targets)."""
    rng = np.random.default_rng(seed)
    pool = _val_pool(train_seqs, train_coords_dict, min_len, max_len); rng.shuffle(pool)
    out = []
    for t_id, seq in pool:
        sub = train_seqs[train_seqs['sequence'] != seq]
        c = get_template_candidates(seq, sub, train_coords_dict, k=1)
        if (not c) or c[0]['ident'] < max_ident: out.append((t_id, seq))
        if len(out) >= n: break
    return out

def tune_pair_prior(val, msa_dir, train_coords_dict, w=None, timeout_s=1800):
    """One DRfold2 pass per validation RNA. Models 1,3,5 are restrained, 2,4 free (UNRESTRAINED_MODELS), so one run
       answers 'does the prior help?'. Returns means of TM-score (model_1, best-of-5, restrained, free) and model RMSD."""
    global PAIR_PRIOR_W
    if w is not None: PAIR_PRIOR_W = w
    free = {int(x) for x in UNRESTRAINED_MODELS.split(',') if x.strip()}
    rows = []
    for t_id, seq in val:
        truth = train_coords_dict[t_id]
        ss = get_consensus_ss(seq, t_id, msa_dir, t_id)
        pairs3 = clean_pairs(parse_pairs_all(ss), seq)
        t0 = time.time()
        models = run_drfold2_models(seq, f"val_{t_id}", t_id, pairs3, timeout_s)
        if not models: continue
        tms = [tm_score_fixed(m, truth) for m in models]
        r = [tms[k] for k in range(len(tms)) if (k + 1) not in free]
        f = [tms[k] for k in range(len(tms)) if (k + 1) in free]
        rm = [rmsd_after_kabsch(a, b) for i, a in enumerate(models) for b in models[i + 1:]]
        rows.append((tms[0], max(tms), np.mean(r) if r else np.nan, np.mean(f) if f else np.nan,
                     np.mean(rm) if rm else np.nan))
        print(f"  {t_id} L={len(seq)} TM m1={tms[0]:.3f} best5={max(tms):.3f} restrained={rows[-1][2]:.3f} "
              f"free={rows[-1][3]:.3f} model-RMSD={rows[-1][4]:.1f}Å ({time.time()-t0:.0f}s)")
    if not rows: return None
    a = np.array(rows, float)
    s = {'w': PAIR_PRIOR_W, 'n': len(rows), 'm1': float(a[:, 0].mean()), 'best5': float(a[:, 1].mean()),
         'restrained': float(np.nanmean(a[:, 2])), 'free': float(np.nanmean(a[:, 3])), 'rmsd': float(np.nanmean(a[:, 4]))}
    print(f"MEAN (w={s['w']}, n={s['n']}): m1={s['m1']:.3f} best-of-5={s['best5']:.3f} "
          f"restrained={s['restrained']:.3f} free={s['free']:.3f} model-RMSD={s['rmsd']:.1f}Å")
    return s

def loo_compare_aligners(train_seqs, train_coords_dict, settings, n=80, min_len=30, max_len=250, seed=0):
    """CPU only. Mean TM of top-1-by-score, top-1-by-identity and best-of-5 templates for each aligner scoring."""
    global TPL_ALIGN
    rng = np.random.default_rng(seed)
    pool = _val_pool(train_seqs, train_coords_dict, min_len, max_len); rng.shuffle(pool); pool = pool[:n]
    res = {}
    for name, params in settings.items():
        TPL_ALIGN = params; _TEMPLATE_CACHE.clear()
        r = []
        for t_id, seq in pool:
            sub = train_seqs[train_seqs['sequence'] != seq]
            cs = get_template_candidates(seq, sub, train_coords_dict, k=5)
            if not cs: r.append((0, 0, 0)); continue
            truth = train_coords_dict[t_id]
            tm = [tm_score_fixed(fill_unmapped(c['raw']), truth) for c in cs]
            r.append((tm[0], tm[int(np.argmax([c['ident'] for c in cs]))], max(tm)))
        a = np.array(r); res[name] = (params, float(a[:, 0].mean()), float(a[:, 1].mean()), float(a[:, 2].mean()))
        print(f"  {name:<20} top1-by-score={a[:,0].mean():.3f}  top1-by-identity={a[:,1].mean():.3f}  best-of-5={a[:,2].mean():.3f}")
    TPL_ALIGN = (2.0, -1.0, -0.5, -0.1); _TEMPLATE_CACHE.clear()
    return res

def budget_table(test_csv, train_seqs, train_coords_dict, caps=(150, 200, 250, 300, 350)):
    df = pd.read_csv(test_csv); lens = []
    for r in df.itertuples():
        try: segs = build_segments(r.all_sequences, r.sequence)
        except Exception: continue
        for s in dict.fromkeys(s for s, is_rna in segs if is_rna):
            if not _template_confident(s, train_seqs, train_coords_dict): lens.append(len(s))
    out = {}
    for cap in caps:
        out[cap] = 1.15 * sum(TM.predict(L) for L in lens if L <= cap) / 3600
        print(f"  CHAIN_MAX_LEN={cap}: {out[cap]:.2f} h of DRfold2 for {sum(L <= cap for L in lens)} chains")
    return out

def thresholds_from_curve(rows, ref_tm):
    """Template identity thresholds from the leave-one-out curve, relative to DRfold2's mean model_1 TM (ref_tm)."""
    rows = np.asarray(rows); edges = [0, 0.3, 0.5, 0.7, 0.8, 0.9, 1.01]
    def first(target, default):
        for lo, hi in zip(edges[:-1], edges[1:]):
            m = (rows[:, 0] >= lo) & (rows[:, 0] < hi)
            if m.sum() >= 3 and rows[m, 1].mean() >= target: return max(lo, 0.3)
        return default
    mid, high, skip = first(0.7 * ref_tm, 0.5), first(ref_tm, 0.8), first(1.15 * ref_tm, 0.9)
    high = max(high, mid); skip = max(skip, high)
    return {'MID_TEMPLATE_ID': round(min(mid, 0.95), 2), 'HIGH_TEMPLATE_ID': round(min(high, 0.95), 2),
            'TEMPLATE_SKIP_CONF': round(min(skip, 0.95), 2)}

def run_tuning(train_seqs, train_coords_dict, test_csv, msa_dir, n_val=6, weights=(0.35, 0.6),
               val_len=(40, 90), budget_h=5.5, n_loo=80, do_prior=True):
    """Everything that can be tuned, in order of cost. Writes /kaggle/working/tuned_config.json and prints the block
       to paste into the config cell. Heuristic decisions are printed with their evidence so you can overrule them."""
    global PAIR_PRIOR_W, TPL_ALIGN, TEMPLATE_RANK_BY
    T = {}
    print("\n[1/5] template aligner + ranking (CPU)")
    res = loo_compare_aligners(train_seqs, train_coords_dict,
                               {'current': (2.0, -1.0, -0.5, -0.1), 'open-2': (2.0, -1.0, -2.0, -0.2),
                                'open-4': (2.0, -1.0, -4.0, -0.5)}, n=n_loo)
    best = None
    for name, (params, t_score, t_ident, t_best5) in res.items():
        for rank, val in (('score', t_score), ('ident', t_ident)):
            if best is None or val > best[0] + 1e-9: best = (val, name, params, rank)
    cur = res['current'][1]
    if best[0] < cur + 0.005: best = (cur, 'current', res['current'][0], 'score')     # keep default unless clearly better
    T['TPL_ALIGN'] = list(best[2]); T['TEMPLATE_RANK_BY'] = best[3]
    TPL_ALIGN = tuple(best[2]); TEMPLATE_RANK_BY = best[3]; _TEMPLATE_CACHE.clear()
    print(f"  -> {best[1]} {best[2]} ranked by {best[3]} (mean TM {best[0]:.3f} vs current {cur:.3f})")

    print("\n[2/5] CHAIN_MAX_LEN from the time budget (CPU)")
    tab = budget_table(test_csv, train_seqs, train_coords_dict)
    ok = [c for c, h in tab.items() if h <= budget_h]
    T['CHAIN_MAX_LEN'] = int(max(ok)) if ok else int(min(tab))
    print(f"  -> CHAIN_MAX_LEN = {T['CHAIN_MAX_LEN']} (largest cap with ≤ {budget_h} h of DRfold2)")

    ref = None
    if do_prior:
        print("\n[3/5] 2D prior strength (GPU, ~1 h)")
        val = pick_validation_set(train_seqs, train_coords_dict, n=n_val, min_len=val_len[0], max_len=val_len[1])
        print("  validation RNAs:", [(t, len(s)) for t, s in val])
        S = [s for s in (tune_pair_prior(val, msa_dir, train_coords_dict, w=w) for w in weights) if s]
        if S:
            bestS = max(S, key=lambda s: s['restrained'])
            gain = bestS['restrained'] - bestS['free']
            T['PAIR_PRIOR_W'] = bestS['w'] if gain >= 0.02 else 0.0
            PAIR_PRIOR_W = T['PAIR_PRIOR_W']
            ref = float(np.mean([s['m1'] for s in S]))
            print(f"  -> PAIR_PRIOR_W = {T['PAIR_PRIOR_W']} (best restrained-free gain {gain:+.3f}, needs ≥ +0.020); "
                  f"DRfold2 mean model_1 TM = {ref:.3f}")
    else:
        print("\n[3/5] skipped (do_prior=False)")

    print("\n[4/5] template identity thresholds (CPU)")
    rows = loo_template_curve(train_seqs, train_coords_dict, n=n_loo)
    if ref is None:
        ref = 0.40; print("  (no DRfold2 reference TM available: assuming 0.40)")
    sug = thresholds_from_curve(rows, ref)
    print("  suggestion only, NOT applied (the reference TM comes from targets WITHOUT templates, so it is biased low;"
          " use RUN_MODE='compare'):", sug)

    print("\n[5/5] pair-geometry centres from the models measured above")
    sc = suggest_constants()
    for k in ('PAIR_PRIOR_CC', 'PAIR_PRIOR_NN'):
        if k in sc: T[k] = list(sc[k])
    print("  ->", {k: T[k] for k in ('PAIR_PRIOR_CC', 'PAIR_PRIOR_NN') if k in T} or "(unchanged)")

    json.dump(T, open('/kaggle/working/tuned_config.json', 'w'), indent=2)
    print("\n════ paste this into the CONFIG cell (TUNED_OVERRIDES) — also saved to /kaggle/working/tuned_config.json ════")
    print("TUNED_OVERRIDES = " + json.dumps(T, indent=4))
    return T

def compare_template_vs_drfold(train_seqs, train_coords_dict, msa_dir, bins=((0.5, 0.7), (0.7, 0.9)),
                               n_per_bin=4, min_len=40, max_len=90, seed=2, timeout_s=1800):
    """Decides when DRfold2 is worth running. Takes training RNAs that DO have a template (best identity inside each
       bin; the target and exact duplicates are removed), folds them with DRfold2 and compares TM-scores of
       template-only, DRfold2-only and the mixed 5-slot set. ~5 min per target on GPU."""
    rng = np.random.default_rng(seed)
    pool = _val_pool(train_seqs, train_coords_dict, min_len, max_len); rng.shuffle(pool)
    chosen = {b: [] for b in bins}
    for t_id, seq in pool:
        if all(len(v) >= n_per_bin for v in chosen.values()): break
        sub = train_seqs[train_seqs['sequence'] != seq]
        cs = get_template_candidates(seq, sub, train_coords_dict, k=5)
        if not cs: continue
        top = max(c['ident'] for c in cs)
        for b in bins:
            if b[0] <= top < b[1] and len(chosen[b]) < n_per_bin:
                chosen[b].append((t_id, seq, sub, top)); break
    print("selected:", {b: [(t, len(s), round(i, 2)) for t, s, _, i in v] for b, v in chosen.items()})
    rows = []
    for b, items in chosen.items():
        for t_id, seq, sub, top in items:
            truth = train_coords_dict[t_id]
            ss = get_consensus_ss(seq, t_id, msa_dir, t_id)
            pairs3 = clean_pairs(parse_pairs_all(ss), seq)
            models = run_drfold2_models(seq, f"cmp_{t_id}", t_id, pairs3, timeout_s)
            if not models: continue
            drf = [{'coords': m, 'name': f'drf{k+1}', 'sat': ss_satisfaction(m, pairs3), 'src': 'drfold', 'conf': 0.0}
                   for k, m in enumerate(models)]
            tmpl = template_models(seq, t_id, msa_dir, sub, train_coords_dict, k=5)
            tm_d = [tm_score_fixed(d['coords'], truth) for d in drf]
            tm_t = [tm_score_fixed(t['coords'], truth) for t in tmpl]
            mix = build_slots([dict(d) for d in drf], [dict(t) for t in tmpl])
            tm_m = [tm_score_fixed(s['coords'], truth) for s in mix]
            rows.append((b, t_id, len(seq), top, tm_d[0], max(tm_d), tm_t[0], max(tm_t), max(tm_m)))
            print(f"  {t_id} L={len(seq)} ident={top:.2f} | DRfold m1={tm_d[0]:.3f} best5={max(tm_d):.3f} | "
                  f"template top1={tm_t[0]:.3f} best5={max(tm_t):.3f} | mixed best5={max(tm_m):.3f}")
    if not rows: return None
    print("\nbin        n  DRfold-m1 DRfold-b5 tmpl-top1 tmpl-b5 mixed-b5")
    summary = {}
    for b in bins:
        r = [x for x in rows if x[0] == b]
        if not r: continue
        a = np.array([x[4:] for x in r], float).mean(0)
        summary[b] = a
        print(f"{b[0]:.1f}-{b[1]:.1f}  {len(r):2d}   {a[0]:.3f}     {a[1]:.3f}    {a[2]:.3f}    {a[3]:.3f}   {a[4]:.3f}")
    print("\nHow to read: if mixed-b5 is not better than tmpl-b5 by ~0.02 in a bin, DRfold2 adds nothing there and you can skip it")
    print("for chains whose best-template identity is at least that bin's lower edge (TEMPLATE_SKIP_CONF). If DRfold-m1 is below tmpl-top1,")
    print("put the template first (HIGH_TEMPLATE_ID). n is tiny (4 per bin): only act on differences larger than ~0.05.")
    return rows
```
